## 2. Key Concepts and Definitions

*   **File Handle**: A temporary link or reference to a file on your system, created when you `open()` a file. In a clinical setting, think of a file handle as the act of pulling a specific patient's chart from the records room; you have it in your hands to read or write on, and you must return it (close it) when you are done. The `with` statement handles this automatically.

*   **File Mode**: A single-character string that tells Python what you intend to do with a file. Each mode corresponds to a common clinical data task.
    *   **Write Mode (`'w'`)**: Creates a new file or completely overwrites an existing one. Analogous to starting a fresh daily report for a hospital ward, discarding the previous day's draft.
    *   **Append Mode (`'a'`)**: Adds new data to the end of an existing file. This is like adding a new entry to a patient's ongoing Medication Administration Record (MAR).
    *   **Read Mode (`'r'`)**: Opens a file for reading only. This is the most common operation, used for reviewing a patient's history, lab results, or admission notes.
    *   **Exclusive Creation Mode (`'x'`)**: Creates a new file but fails with an error if the file already exists. This is a safety measure, perfect for creating a unique, timestamped log file that should never be accidentally overwritten.

*   **Text File**: A simple file format that stores data as human-readable characters. Patient notes, lab reports, and configuration files are often stored as text files. We specify `encoding='utf-8'` to ensure compatibility with a wide range of characters.

*   **Exception Handling**: A mechanism using `try...except` blocks to manage errors during program execution without crashing. In a medical application, this is non-negotiable. If a patient file is missing, the program should report the error and continue running, not shut down the entire system.

## 3. Main Content

### 3.1. Writing to a New File (`'w'` mode)

The `'w'` mode creates a new file for writing.

> **Important:** Using `'w'` mode on an existing file will permanently erase all of its contents. This action cannot be undone.

> **In Practice:** This mode is ideal for tasks like generating a new, unique report file, such as a discharge summary for a specific patient encounter (e.g., `summary_PT004598_20251111.txt`).

````python
discharge_summary = "Patient discharged in stable condition. Follow up in 2 weeks."
file_path = 'summary_report.txt'
with open(file_path, 'w', encoding='utf-8') as summary_file:
    summary_file.write(discharge_summary)
print(f"Successfully wrote to {file_path}")
````

### 3.2. Appending to an Existing File (`'a'` mode)

Append mode (`'a'`) adds data to the end of a file without erasing existing content.

> **In Practice:** Append mode is essential for maintaining running logs, such as a patient's Medication Administration Record (MAR) or continuous vital signs monitoring, where each new entry is added chronologically without altering past data.

````python
# Assume 'patient_mar.txt' already contains "09:00 - Administered aspirin 81mg PO."
new_log_entry = "\n09:15 - Administered metformin 500mg PO."
file_path = 'patient_mar.txt'
with open(file_path, 'a', encoding='utf-8') as log_file:
    log_file.write(new_log_entry)
print(f"Successfully appended to {file_path}")
````

### 3.5. Integrated Example: Filtering Critical Lab Notes

This script reads `lab_notes.txt`, filters for lines containing "CRITICAL", and writes them to `critical_alerts.txt`.

> **Pro Tip:** This example provides a preview of two powerful string methods, `.upper()` and `.startswith()`, which you will master in sub-chapter 1.11. They allow for flexible and case-insensitive text matching.

````python
# --- Setup: Create the lab notes file ---
# Example line in lab_notes.txt: "CRITICAL: Potassium level low (2.9 mmol/L). Notify provider immediately."
notes = [
    "Glucose: 150 mg/dL - HIGH",
    "CRITICAL: Potassium level is 2.1 mEq/L. Immediate action required.",
    "WBC: 11.5 x10^9/L",
    "CRITICAL: Troponin level elevated. Possible cardiac event."
]
with open('lab_notes.txt', 'w', encoding='utf-8') as f:
    f.write('\n'.join(notes))
# -----------------------------------------

CRITICAL_KEYWORD = "CRITICAL"
critical_count = 0
try:
    with open('lab_notes.txt', 'r', encoding='utf-8') as notes_file, \
         open('critical_alerts.txt', 'w', encoding='utf-8') as alerts_file:
        for line in notes_file:
            if line.strip().upper().startswith(CRITICAL_KEYWORD):
                alerts_file.write(line)
                critical_count += 1
        
        if critical_count == 0:
            alerts_file.write("No critical notes found.")

    print(f"Processing complete. {critical_count} alerts written to critical_alerts.txt.")

except FileNotFoundError:
    print("Error: 'lab_notes.txt' not found.")
````

## 5. Practical Applications

*   **Electronic Health Record (EHR) Data Logging:** Every patient interaction, from physician notes to medication administration, must be logged. Python scripts can use append mode (`'a'`) to add timestamped entries to a patient's master log file. This creates a secure, chronological, and auditable record of care, which is fundamental to modern healthcare.

*   **Generating Daily Clinical Reports:** Hospitals need to summarize vast amounts of data daily. A Python program can be scheduled to run every night, reading data from dozens of files (e.g., `lab_results.txt`, `vitals.txt`, `admissions.txt`) and using write mode (`'w'`) to generate a single, consolidated `summary_report_YYYY-MM-DD.txt`. This automates a critical administrative task, providing clinicians with a concise overview for their morning rounds.

*   **Genomic Data Processing Pipelines:** Genomic files (like FASTA or VCF formats) can be gigabytes in size. Loading such a file entirely into memory with `.read()` is impossible. The line-by-line iteration technique (`for line in file:`) is essential here. Bioinformaticians use this method to build pipelines that read these massive files piece by piece, filtering for specific genes or mutations and writing results to a new file, enabling large-scale analysis on standard hardware.

### <summary>Solution</summary>

````python
# 1. Setup: Create the log file
hr_data = "72\n75\n68\n81\n78"
with open('hr_log.txt', 'w', encoding='utf-8') as f:
    f.write(hr_data)

# 2. Read the file and calculate the average
total_hr = 0
count = 0
try:
    with open('hr_log.txt', 'r', encoding='utf-8') as f:
        for line in f:
            total_hr += int(line.strip())
            count += 1
    
    if count > 0:
        average_hr = total_hr / count
        print(f"Average heart rate: {average_hr:.2f}")
    else:
        print("Log file is empty.")

except FileNotFoundError:
    print("Error: 'hr_log.txt' not found.")
except ValueError:
    print("Error: Log file contains non-numeric data.")
````

### <summary>Solution</summary>

````python
# Setup: Create the source notes file
notes_content = "14:30 - Patient stable.\n15:00 - URGENT: Patient reporting severe headache.\n16:00 - URGENT: Blood pressure drop detected."
with open('all_notes.txt', 'w', encoding='utf-8') as f:
    f.write(notes_content)

urgent_notes = []
# First, try to read the source file and collect urgent notes
try:
    with open('all_notes.txt', 'r', encoding='utf-8') as source_file:
        for line in source_file:
            if "URGENT" in line:
                urgent_notes.append(line)

    # If urgent notes were found, try to write them to the archive file
    if urgent_notes:
        try:
            with open('urgent_archive.txt', 'x', encoding='utf-8') as archive_file:
                archive_file.writelines(urgent_notes)
            print(f"Successfully archived {len(urgent_notes)} urgent notes.")
        except FileExistsError:
            print("Error: 'urgent_archive.txt' already exists. Archiving aborted to prevent data loss.")
    else:
        print("No urgent notes found in 'all_notes.txt'.")

except FileNotFoundError:
    print("Error: Source file 'all_notes.txt' not found.")
````
