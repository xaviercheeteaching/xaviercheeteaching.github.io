## 2. Key Concepts and Definitions

*   **Pandas DataFrame**: A two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). In a medical context, this is the digital equivalent of a patient cohort spreadsheet, where each row represents a unique patient and each column represents a specific measurement or attribute (e.g., age, blood pressure, diagnosis).
*   **Attribute**: A characteristic or property of a DataFrame that provides metadata without performing an action or calculation. For example, `df.shape` is an attribute that tells you the dimensions of your patient data (number of patients, number of variables) without altering the data itself.
*   **Method**: A function that belongs to a DataFrame object and performs an action, such as a calculation or data transformation. For instance, `df.head()` is a method that actively retrieves and displays the first few rows of your dataset, like quickly glancing at the first few patient files in a stack.
*   **Index**: The label for the rows in a DataFrame. By default, this is a sequence of integers starting from 0 (e.g., 0, 1, 2, ...), acting like the sequential chart numbers assigned to patients as they are admitted to a study.

> **Important:** Pay close attention to syntax: **attributes** like `df.shape` do not have parentheses `()`, while **methods** like `df.head()` require them. Forgetting the parentheses for a method is a very common error.

### 3.1 Creating a DataFrame

Dr. Chen has provided you with a small dataset of new admissions at City Medical Center. Let's load it into a DataFrame. The most common way to do this is by using a Python dictionary, where each key becomes a column header and the list of values becomes the data for that column.

````python
import pandas as pd

# Define the data using snake_case for columns and string IDs
patient_data = {
    'patient_id': ['PT000101', 'PT000102', 'PT000103', 'PT000104'],
    'age': [45, 62, 38, 55],
    'blood_type': ['A+', 'O-', 'B+', 'A+'],
    'systolic_bp_mmhg': [120, 145, 115, 130]
}
df_patients = pd.DataFrame(patient_data)

print(df_patients)
````

### 3.2 Inspecting DataFrame Properties with Attributes

Attributes are like asking your DataFrame for its ID card—they give you quick facts about its structure without changing it.

````python
import pandas as pd

# Data setup from the previous example
patient_data = {
    'patient_id': ['PT000101', 'PT000102', 'PT000103', 'PT000104'],
    'age': [45, 62, 38, 55],
    'blood_type': ['A+', 'O-', 'B+', 'A+'],
    'systolic_bp_mmhg': [120, 145, 115, 130]
}
df_patients = pd.DataFrame(patient_data)

# Inspection code
print(df_patients.shape)
print(df_patients.columns)
print(df_patients.index)
print(df_patients.dtypes)
````

### 3.3 Performing Actions with Methods

Methods are commands that ask the DataFrame to do something, like show you a summary or a preview of the data.

````python
import pandas as pd

# Data setup from the previous example
patient_data = {
    'patient_id': ['PT000101', 'PT000102', 'PT000103', 'PT000104'],
    'age': [45, 62, 38, 55],
    'blood_type': ['A+', 'O-', 'B+', 'A+'],
    'systolic_bp_mmhg': [120, 145, 115, 130]
}
df_patients = pd.DataFrame(patient_data)

# Get a concise technical summary
df_patients.info()

# Show a preview of the first 2 rows
print(df_patients.head(2))

# Show a preview of the last 2 rows
print(df_patients.tail(2))
````

## 5. Practical Applications

*   **Clinical Trial Data Management**: In clinical trials, researchers collect diverse data points (demographics, dosage, vitals, outcomes) for hundreds of participants. A Pandas DataFrame is the standard tool to organize this data. The `.shape` attribute gives an instant overview of the trial's scale (participants vs. variables), while `.info()` helps identify missing data points that need to be addressed.
*   **Electronic Health Record (EHR) Analysis**: Hospitals aggregate patient data from EHRs into DataFrames to monitor trends. An analyst might pull hourly vital signs for all ICU patients into a DataFrame. Using `.head()` and `.tail()` on this time-series data allows them to quickly check the most recent and oldest records, which is crucial for identifying acute changes in a patient's condition.
*   **Genomic Data Structuring**: In genomics, gene expression data is often represented in a large matrix where rows are genes and columns are different patient samples. Loading this matrix as a DataFrame allows bioinformaticians to immediately use `.dtypes` to ensure expression values are numeric and `.columns` to verify that all sample IDs are correctly loaded, forming the basis for identifying genetic markers for disease.
