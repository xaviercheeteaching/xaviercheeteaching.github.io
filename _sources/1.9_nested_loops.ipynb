# Chapter 1: Control Flow and Looping

## 1. Introduction

This section introduces a powerful programming structure: the nested `while` loop. You will learn how to place one `while` loop inside another to perform repetitive, multi-step tasks efficiently. The learning journey will focus on understanding how an "inner" loop can run a complete sub-process for each single step of an "outer" loop. We will use a clinical analogy of titrating medication for a series of patients to demonstrate how this structure is essential for handling complex, layered procedures in health data simulations.

> **In Practice:** In automated infusion systems or clinical simulations, a process for one patient (like titrating a drug to a target blood pressure) often needs to be repeated for a list of patients. A nested loop is the perfect structure for this, where the outer loop iterates through patients and the inner loop handles the complete titration protocol for each one.

---

## 2. Key Concepts and Definitions

*   **Nested Loop**: A control flow structure where one loop is placed inside the body of another. In a medical context, this is like running a full diagnostic protocol (the inner loop) for every patient admitted to a ward (the outer loop).
*   **Outer Loop**: The primary, enclosing loop that controls the overall number of major iterations. For example, an outer loop might iterate through a list of `patient_id`s.
*   **Inner Loop**: The secondary, enclosed loop that executes completely for *each* single iteration of the outer loop. For instance, an inner loop could simulate the minute-by-minute adjustment of a drug infusion until a patient's vital sign stabilizes.
*   **State Reset**: The process of re-initializing variables at the beginning of the outer loop's block. This is critical to ensure that the inner loop starts with fresh, predictable conditions for each new item being processed (e.g., resetting a patient's simulated blood pressure to its initial value before starting titration).
*   **Guaranteed Exit**: A logical test that determines when a loop should terminate. In a nested structure, both the inner and outer loops must have their own independent, guaranteed exit conditions to prevent the program from getting stuck in an infinite loop.

---

## 3. Main Content

A nested `while` loop executes a complete inner loop for each single iteration of an outer loop. This is used to apply a multi-step procedure, like titrating a medication, to a series of distinct patients. In this simulation, each rate increase causes an immediate drop in blood pressure; a real clinical response is more gradual and variable.

The outer loop will manage the patients, and for each patient, the inner loop will handle the titration process until that specific patient's blood pressure target is met.

```python
patient_number = 1
# Outer loop: iterates through each patient
while patient_number <= 2:
    patient_id = f"PT00000{patient_number}" # Standardized ID for display
    # State Reset: Reset variables for the new patient's simulation
    systolic_bp = 180
    infusion_rate = 0  # Antihypertensive in mL/hr
    print(f"\n--- Titrating Patient {patient_id} (Initial Systolic Blood Pressure (BP): {systolic_bp} mmHg) ---")

    # Inner loop: runs until the BP target is met for the current patient
    while systolic_bp > 130:
        infusion_rate += 1
        systolic_bp -= 5  # Simulate BP reduction from drug
        print(f"  Rate: {infusion_rate} mL/hr -> Systolic BP: {systolic_bp} mmHg")
    
    print(f"  Target BP reached.")
    patient_number += 1 # Update outer loop counter
```

> **Debug Note:** A nested loop has two common failure points. First, forgetting to reset the inner loop's variables (`systolic_bp`, `infusion_rate`) for each run of the outer loop. This causes the inner loop to run correctly only for the first patient. Second, failing to include an update step (`patient_number += 1` and `systolic_bp -= 5`) for *each* loop's condition, which will cause an infinite loop.

    plate_id += 1
```
**Explanation:** The outer loop handles the three plates. For each plate, `day` and `confluency` are reset. The inner loop models the growth, incrementing both `confluency` and `day` until the 80% threshold is met. The final message uses `day - 1` because the `day` variable is incremented one final time just before the loop condition becomes false.

**Key Learning:** Nested loops are effective for running time-based simulations across multiple subjects or experiments.

patient_num = 1
while patient_num <= 2:
    # Reset state for the new patient
    time_elapsed = 0
    workload = 2.0
    current_hr = 80
    max_hr_target = 160
    max_duration = 15
    
    print(f"\n--- Starting Stress Test for Patient {patient_num} ---")
    
    # Inner loop with multiple exit conditions
    while current_hr < max_hr_target and time_elapsed < max_duration:
        print(f"  Time: {time_elapsed} min, Workload: {workload:.1f} METs, HR: {current_hr} bpm")
        time_elapsed += 1
        workload += 0.5
        current_hr += 10
        
    # Check which condition caused the loop to terminate
    if current_hr >= max_hr_target:
        print(f"Test complete: Target heart rate of {max_hr_target} bpm reached at {time_elapsed} minutes.")
    else:
        print(f"Test complete: Maximum duration of {max_duration} minutes reached.")
        
    patient_num += 1
```
**Explanation:** The inner loop's `while` condition uses `and` to ensure it stops as soon as *either* the heart rate or time limit is reached. After the loop, an `if/else` statement checks the final value of `current_hr` to determine which condition was met and prints the appropriate summary.

**Key Learning:** Inner loops can have complex exit conditions involving multiple variables, which is common in clinical protocols with several safety cutoffs.

## 5. Practical Applications

Nested loops are fundamental for processing hierarchical or multi-dimensional data, a common task in precision health.

*   **Genomic Variant Analysis:** A program might use an outer loop to iterate through a list of patients. For each patient, an inner loop iterates through thousands of their genetic variants, comparing each one against a database of known pathogenic mutations. This allows for automated screening of entire cohorts for specific disease-causing genes.
*   **Pharmacokinetic/Pharmacodynamic (PK/PD) Modeling:** Researchers can use an outer loop to test different drug compounds. For each compound, an inner loop simulates its effect over time at various dosages, calculating metrics like absorption, distribution, metabolism, and excretion. This helps identify the most promising drug candidates before clinical trials.
*   **Medical Image Batch Processing:** In radiology, an outer loop can iterate through a directory of patient CT scans. For each scan (which consists of many image slices), an inner loop processes every slice to perform tasks like identifying tumors, measuring organ volume, or detecting hemorrhages. This automates a tedious manual process and ensures consistency.

## 6. Summary and Key Takeaways

In this section, we've explored how to construct and use nested `while` loops to execute a complete sub-process for each item in a larger set. This pattern is crucial for automating complex, repetitive workflows in computational medicine.

*   A nested loop consists of an outer loop and an inner loop; the inner loop runs to completion for every single iteration of the outer loop.
*   Resetting the state of inner loop variables at the start of each outer loop iteration is essential to prevent data from one iteration from corrupting the next.
*   Both the inner and outer loops must have clear and independent exit conditions to avoid infinite loops.
*   This structure is highly applicable in scenarios like running simulations for multiple patients, analyzing layered data (e.g., genes within patients), or processing batches of complex files (e.g., slices within medical images).
