## 2. Key Concepts and Definitions

Understanding these control statements is key to writing efficient and readable loops.

*   **Loop Control Statement**: A command that alters the normal execution flow of a loop. In a clinical system, these are used to react to specific data points, like stopping a search for a genetic marker once it's found or skipping over records that don't need processing.
*   **`break`**: This statement immediately terminates the innermost loop it is in. Think of it as a clinical "stat" order that stops all other processing to focus on a critical finding, such as halting a review of patient vitals once a value crosses a life-threatening threshold.
*   **`continue`**: This statement skips the rest of the code inside the current iteration of a loop and proceeds to the next iteration. This is useful for filtering data, similar to a lab technician skipping over stable samples to focus only on those that are flagged for further analysis.
*   **`for...else` Clause**: The `else` block attached to a `for` loop executes only when the loop has completed all its iterations naturally, without being terminated by a `break` statement. This is a powerful way to confirm a condition's absence, such as verifying that a patient has no recorded allergies to a prescribed medication.

### 3.1 The `break` Statement: Exiting a Loop Early

The `break` statement provides a way to exit a loop immediately. This is highly efficient for "search" operations where you only need to find the first occurrence of a condition.

**Use Case:** Find the first heart rate indicating tachycardia in a log. Once found, flag it and stop checking the rest of the log.

````python
# In a real application, thresholds are often configured as constants.
ALERT_THRESHOLD = 100
heart_rate_log = [72, 75, 81, 105, 88, 92]

for heart_rate in heart_rate_log:
    if heart_rate > ALERT_THRESHOLD:
        print(f"Heart rate {heart_rate} bpm exceeds threshold. Flagging for review.")
        break
    print(f"Heart rate stable: {heart_rate} bpm.")
````

Using `break` ensures the system can issue an alert immediately without wasting precious seconds or milliseconds processing non-critical data that follows.

### 3.2 The `continue` Statement: Skipping an Iteration

The `continue` statement allows you to skip the remainder of the current iteration and move directly to the next one. It is ideal for filtering out items you don't need to process.

**Use Case:** Process only core lab panels, such as a CBC (Complete Blood Count) or BMP (Basic Metabolic Panel), from a list of test orders.

````python
lab_orders = ["CBC", "LIPID", "BMP", "TSH"]
core_panels = ["CBC", "BMP"]

for test in lab_orders:
    if test not in core_panels:
        continue  # Skip non-core tests
    print(f"Processing core lab order: {test}")
````

> **Pro Tip:** The `continue` statement is a powerful tool for data cleaning and validation. Before performing complex calculations or analysis on a dataset, you can loop through it and use `continue` to skip over rows with missing values, invalid entries, or data points that fall outside the scope of your current task.

### 3.3 The `else` Clause in a For Loop: The "No Break" Clause

The `else` clause runs only if the loop completes all its iterations without a `break`. Think of it as a 'no break occurred' block, perfect for confirming an item was not found.

**Use Case:** Check a patient's recorded allergies before administering a drug.

````python
patient_allergies = ["Penicillin", "Sulfa", "Ibuprofen"]
DRUG_TO_ADMINISTER = "Aspirin"

for allergy in patient_allergies:
    if allergy == DRUG_TO_ADMINISTER:
        print(f"Allergy found: {allergy}. Do not administer.")
        break
else:
    # Executes only because the loop finished without a 'break'.
    print(f"No allergy to {DRUG_TO_ADMINISTER} found. Proceed with safety protocols.")
````

> **Pro Tip:** Think of the `else` on a loop as the "all-clear" signal. It only runs if the loop completed its full patrol (all iterations) without raising an alarm (`break`). This makes it perfect for confirming that a specific condition was *not* met in an entire dataset.

> **Debug Note:** A common mistake is confusing a loop's `else` clause with a standard `if/else` statement. Remember, the loop's `else` is tied to the `break` statement, not the `if` condition. It will *not* run if the loop is exited via `break`.

## 5. Practical Applications

*   **Real-time Vitals Monitoring:** An ICU monitoring system can loop through a patient's vital signs. Using `break`, the system can immediately trigger an alarm and halt the routine check the instant any single vital sign crosses a critical threshold, ensuring the most urgent issue is addressed without delay.
*   **Genomic Data Filtering:** When analyzing a patient's genomic sequence, researchers might search for a list of specific genetic markers. Using `continue`, a script can iterate through millions of base pairs, skipping over all non-marker sequences and only processing the ones that are relevant to the study, dramatically improving efficiency.
*   **Clinical Trial Eligibility Screening:** A program can screen a patient's electronic health record to determine eligibility for a clinical trial. It can loop through a list of disqualifying conditions (e.g., specific allergies). If the loop completes without finding any disqualifying conditions (`break` is never called), the `else` block can run to flag the patient as "Potentially Eligible," confirming they passed the initial screening.
