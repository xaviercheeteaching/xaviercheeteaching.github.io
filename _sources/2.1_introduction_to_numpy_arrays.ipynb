## 2. Key Concepts and Definitions

- **Array**: A data structure that stores a collection of items of the same data type (homogeneous). This uniformity is what allows NumPy to perform mathematical operations on the entire collection at once, a process called vectorization, which is significantly faster than looping through a Python list. In medicine, an array could efficiently store thousands of continuous glucose monitoring readings.

- **`.shape`**: A tuple representing the array's dimensions. For a 2D array, it's `(rows, columns)`, like a spreadsheet. For an array of patient vitals, a shape of `(3, 3)` means 3 patients (rows) and 3 vital signs (columns).

- **`.ndim`**: The number of dimensions or axes. A 1D array (like a single patient's heart rate over time) has `1` dimension, while a 2D array or matrix (like vitals for multiple patients) has `2`.

- **`.size`**: The total number of elements in the array. It's the product of the dimensions in `.shape` (e.g., for a `(3, 3)` array, the size is 3 * 3 = 9).

- **`.dtype`**: The data type of the elements (e.g., `int64`, `float64`). Knowing the `dtype` is crucial for memory management and preventing calculation errors.

- **Type Up-casting**: The process where NumPy automatically converts all elements to the most precise data type when an array is created from mixed-type data. For example, if a list of glucose readings contains one float (`115.5`), all integer readings are converted to floats to prevent data loss.

### 3.1 Creating Arrays from Lists

The most direct way to create an array is by converting a Python list or list of lists using `np.array()`.

````python
import numpy as np

# Hourly post-meal glucose readings (mg/dL)
glucose_readings = [110, 115, 120, 118, 122]
np_glucose = np.array(glucose_readings)

print(np_glucose)

# Expected Output:
# [110 115 120 118 122]
````

### 3.2 Inspecting Array Attributes

After creating an array, use its attributes to verify its structure. This is a critical step for debugging and data validation.

````python
import numpy as np

# Vitals for 3 patients: [Heart Rate (bpm), Systolic Blood Pressure (mmHg), Oxygen Saturation (%)]
patient_vitals = [
    [72, 120, 98],
    [88, 125, 96],
    [65, 110, 99]
]
np_vitals = np.array(patient_vitals)

# Inspecting the 2D np_vitals array
print("Shape:", np_vitals.shape)
print("Dimensions:", np_vitals.ndim)
print("Size:", np_vitals.size)
print("Data Type:", np_vitals.dtype)

# Expected Output:
# Shape: (3, 3)
# Dimensions: 2
# Size: 9
# Data Type: int64
````

### 3.4 Creating Placeholder Arrays

It's often efficient to create placeholder arrays of a desired shape with `np.zeros()` or `np.ones()` and fill them with data later. This pre-allocates the necessary memory.

> **Pro Tip:**
> Pre-allocating an array with `np.zeros()` or `np.ones()` is a professional best practice. It reserves a contiguous block of memory upfront, which is more efficient than incrementally growing a data structure. This is especially important when logging large amounts of data from medical devices in real-time.

````python
import numpy as np

# Create a 24x3 array of zeros for hourly [HR, SBP, SpO2]
# Initializes as float64 by default, a good choice for measurements that might require decimals.
# In a real Electronic Health Record (EHR), some of these values (like Heart Rate) might be stored as integers, so choosing the right dtype is key.
vitals_log = np.zeros((24, 3))
print("Vitals Log Placeholder:")
print(vitals_log)

# Create a 2x2 array of ones to represent pre-op checklist status (1 = complete)
task_status = np.ones((2, 2), dtype=np.int8)
print("\nTask Completion Status:")
print(task_status)

# Expected Output:
# Vitals Log Placeholder:
# [[0. 0. 0.]
#  [0. 0. 0.]
#  ...
#  [0. 0. 0.]]
#
# Task Completion Status:
# [[1 1]
#  [1 1]]
````

### 3.5 Managing Data Types

Controlling the data type of your array is key to managing memory and precision.

> **Important:**
> Choosing the correct `dtype` is a trade-off between memory and precision. Using `np.float64` ensures high precision for sensitive measurements but uses more memory. Using `np.int8` is very memory-efficient for whole numbers between -128 and 127 (like fall counts), but would be inappropriate for storing body temperature or lab results with decimals. Mismatched `dtype`s can lead to data loss or errors in clinical calculations.

````python
import numpy as np

# Drug doses in milligrams (mg)
doses_mg = [50, 100, 75]
np_doses = np.array(doses_mg, dtype=np.float64)

print(np_doses)
print(np_doses.dtype)

# Expected Output:
# [ 50. 100.  75.]
# float64
````

NumPy's automatic up-casting is helpful but can be a silent source of bugs.

> **Debug Note:**
> Automatic type up-casting is helpful but can be a silent source of bugs. If you expect an array of integers but accidentally include one float, the entire array becomes float. This can increase memory usage unexpectedly or cause issues with systems that require integer inputs. Always double-check `.dtype` after creating an array from mixed-type data.

````python
import numpy as np

# A mix of integer and float glucose readings (mg/dL)
mixed_glucose = [110, 115.5, 120]
np_mixed_glucose = np.array(mixed_glucose)

print(np_mixed_glucose)
print(np_mixed_glucose.dtype)

# Expected Output:
# [110.  115.5 120. ]
# float64
````

## 5. Practical Applications

- **Genomic Sequence Analysis**: A human genome contains billions of base pairs. Storing this as a NumPy array of integers (e.g., A=0, C=1, G=2, T=3) using `np.array()` is vastly more memory-efficient and faster to process than Python lists. This allows researchers to perform large-scale pattern matching and mutation detection.

- **Medical Image Processing**: An MRI or CT scan is fundamentally a 2D or 3D array of pixel intensities. NumPy is used to create and manipulate these images. For example, a data scientist might use `np.zeros()` to create an array of the same shape as an MRI scan to serve as a "mask" for isolating a tumor.

- **Real-Time Patient Monitoring**: Wearable sensors like ECGs or continuous glucose monitors generate streams of data. This time-series data is naturally stored in 1D NumPy arrays. `np.array()` can be used to capture incoming data chunks, and functions like `np.linspace()` can help in resampling data to a standard time interval for consistent analysis.
