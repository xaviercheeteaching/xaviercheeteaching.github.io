## 1. Introduction

A `for` loop is a fundamental programming tool that automates repetitive tasks. In precision health, you will frequently need to perform the same action on many pieces of data—for example, checking every temperature reading in a patient's chart or analyzing each base in a DNA sequence. Performing these tasks manually is slow and prone to error. The `for` loop makes this process efficient, scalable, and reliable by executing a block of code for each item in a sequence.

> **In Practice:** In a hospital setting, `for` loops are workhorses for automation. They can be used to process thousands of patient lab results from a list, check an entire medication regimen for potential drug interactions, or generate daily administrative reports—all without manual intervention, ensuring both speed and accuracy.

This section will teach you how to construct `for` loops to process sequences like strings and lists, a foundational skill for any data analysis task in a clinical or research setting.

---

## 2. Key Concepts and Definitions

Understanding the following terms is essential for writing and debugging `for` loops.

*   **Sequence**: An ordered collection of items. A DNA sample represented as the string `"AGCT"` or a list of patient ages `[45, 62, 38]` are both sequences that a `for` loop can process.
*   **Iterable**: Any Python object that can be looped over. Lists and strings are common iterables. An electronic health record system might return a list of a patient's prescribed medications, which is an iterable you can process to check for interactions.
*   **Iteration**: A single pass through the loop's code block, where one item from the sequence is processed. During a loop that checks patient temperatures, a single iteration would involve evaluating one temperature reading.
*   **Loop Variable**: A temporary variable that holds the current item from the sequence during each iteration. In the code `for base in dna_sequence:`, the `base` variable holds a different character value in each iteration of the loop.

---

## 3. Main Content

A `for` loop provides a clean and readable way to iterate over the items of any sequence. We will explore the two most common use cases: iterating over strings and lists.

### 3.1 Iterating Over a String

A `for` loop can process each character in a string one by one. This is a fundamental operation in bioinformatics for analyzing sequential data like a genetic code.

````python
# A string representing a short DNA sequence
dna_sequence = "AGCT"

# The loop variable 'base' will hold one character from the string in each iteration
for base in dna_sequence:
    print(base)

# Expected Output:
# A
# G
# C
# T
````

### 3.2 Iterating Over a List

Similarly, a `for` loop can process each element in a list. This is ideal for working with collections of data, such as a set of patient vital signs or a list of lab results.

````python
# A constant to define the clinical threshold for fever
FEVER_THRESHOLD_CELSIUS = 38.0

# A list of a patient's daily temperature readings in Celsius
daily_temps_celsius = [37.5, 38.1, 37.9, 39.2] 

# The loop variable 'temp' holds one temperature from the list in each iteration
for temp in daily_temps_celsius:
    if temp >= FEVER_THRESHOLD_CELSIUS:
        print(f"Fever detected: {temp}°C")

# Expected Output:
# Fever detected: 38.1°C
# Fever detected: 39.2°C
````

```python
        gc_count += 1 # Short-hand for gc_count = gc_count + 1

# Avoid division by zero if the string is empty

    gc_content_percentage = (gc_count / total_length) * 100
    print(f"GC-Content: {gc_content_percentage:.2f}%")
else:
    print("DNA fragment is empty.")

# Expected Output:

```
**Explanation:** The loop iterates through each `base` in the `dna_fragment`. An `if` statement checks if the base is either 'G' or 'C'. If it is, `gc_count` is incremented. After the loop, the GC-content percentage is calculated and printed.

## 5. Practical Applications

`For` loops are a workhorse in real-world precision health applications, enabling the automation of data processing at scale.

*   **Genomic Sequence Analysis:** In bioinformatics, `for` loops are used to iterate through DNA or RNA sequences (strings). This programming technique allows researchers to find specific patterns (gene motifs), calculate nucleotide frequencies (like GC-content), and identify mutations, which has a real-world impact on disease research and genetic diagnostics.
*   **Batch Processing of Electronic Health Records (EHR):** A `for` loop can process a list of thousands of patient records retrieved from a database. This allows a program to check for drug contraindications, flag patients due for a screening, or anonymize data for a clinical study, improving patient safety and research efficiency.
*   **Analyzing Time-Series Sensor Data:** Data from wearables like smartwatches is collected as a list of vital signs. A `for` loop can iterate through these readings to detect anomalies (e.g., a sudden spike in heart rate), calculate daily averages, or identify trends, providing critical insights for remote patient monitoring.
