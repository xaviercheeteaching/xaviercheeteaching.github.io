## 1. Introduction

Welcome to the core of clinical logic in programming. In this section, you will learn how to use comparison operators to ask questions about medical data. These operators evaluate conditions—such as whether a patient's vital signs are within a normal range—and return a simple `True` or `False` answer. This journey will establish the fundamental skill of creating logical expressions, which is the first step in building automated clinical decision-making tools. By the end, you will be able to use comparison operators to evaluate medical data and return a Boolean result.

> **Important: Comparison (`==`) vs. Assignment (`=`)**
> A single equal sign (`=`) is used to *assign* a value to a variable (e.g., `patient_age = 65`). A double equal sign (`==`) is used to *compare* two values (e.g., `patient_age == 65`). Mixing these up is a very common bug for new programmers.

---

## 2. Key Concepts and Definitions

Understanding the following terms is crucial for using comparison operators effectively in a medical context.

*   **Comparison Operator**: A symbol (like `>`, `==`, or `!=`) used to compare two values. In healthcare programming, these are the tools we use to check if a patient's lab result, vital sign, or age meets a specific clinical criterion.
*   **Boolean Value**: A data type that can only be one of two values: `True` or `False`. A Boolean is the digital answer to a clinical yes-or-no question, such as "Does the patient have a fever?" (`True`) or "Is the patient's blood pressure in the normal range?" (`False`).
*   **Operand**: The value or variable on either side of an operator. In the expression `patient_age >= 65`, `patient_age` and `65` are the operands.
*   **Expression**: A combination of values, variables, and operators that Python evaluates to produce a result. The expression `systolic_bp_mmhg < 90` evaluates to a single Boolean value (`True` or `False`), indicating whether a patient is hypotensive.

> A common bug in data processing is comparing a number to a string representation of that number (e.g., `90 == '90'`). This will always evaluate to `False` and can lead to incorrect logic. Always ensure data types are consistent before comparison.

````python
systolic_bp_num = 90
systolic_bp_str = '90'
print(f"Is {systolic_bp_num} == '{systolic_bp_str}'? {systolic_bp_num == systolic_bp_str}")
# Output: Is 90 == '90'? False
````

## 5. Practical Applications

Comparison operators are actively used in real-world health technology to improve patient safety and outcomes.

*   **Clinical Decision Support (CDS) Systems:** Hospital electronic health record (EHR) systems use comparison operators to power real-time alerts. For example, a CDS rule might continuously check `if patient_temp_c > 38.0 and heart_rate_bpm > 90`. If this expression evaluates to `True`, it can trigger a sepsis screening alert for the clinical team.
*   **Automated Patient Triage Algorithms:** In busy emergency departments, software uses expressions like `systolic_bp_mmhg < 90` or `patient_age >= 75` to assess risk factors. The `True`/`False` results from these comparisons help categorize patients by acuity level, ensuring that the most critical cases are seen first.
*   **Medication Dosing and Safety Checks:** Pharmacy information systems embed comparison logic to prevent errors. When a pharmacist enters a prescription, the system checks if `calculated_dose_mg <= MAXIMUM_SAFE_DOSE_MG`. If this returns `False`, the system flags the prescription for review, preventing a potential overdose.
