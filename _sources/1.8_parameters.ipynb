## 1. Introduction

This section introduces one of the most fundamental concepts in programming: functions. In the context of precision health, functions are the building blocks for creating reliable and reusable medical logic. Think of a function as a standardized clinical protocol that you can execute on demand—whether it's for calculating a patient's BMI, determining a drug dosage, or converting temperature units. Our learning journey will begin with defining simple functions that return a fixed value. We will then progress to making them more flexible by adding parameters, explore different ways to provide arguments, and finally, learn how to handle invalid inputs gracefully to ensure our clinical applications are both powerful and safe.

## 2. Key Concepts and Definitions

*   **Function**: A named, reusable block of code that performs a specific task. In a medical setting, a function is like a defined protocol for a calculation, such as determining a patient's Mean Arterial Pressure (MAP) from their blood pressure readings.
*   **Parameter**: The variable name listed inside a function's parentheses (e.g., `weight_lbs`). It acts as a placeholder for the input data the function needs to operate.
*   **Argument**: The actual value that is passed to a function when it is called (e.g., `154.0`). This is the real data, like a specific patient's weight, that fills the parameter's placeholder.
*   **Return Value**: The output that a function sends back after it has completed its task. A function to check for fever might return a boolean (`True` or `False`), while a dosage calculator would return a number (a float or integer).
*   **Docstring**: A string literal at the beginning of a function that explains what it does, its arguments, and what it returns. It is crucial for creating understandable and maintainable code, especially in collaborative clinical research projects.
*   **Type Hint**: A notation that suggests the expected data type of a variable, parameter, or a function's return value (e.g., `-> float`). It helps with code clarity and allows tools to detect potential errors, which is critical for safety.
*   **Scope**: The region of code where a variable is accessible. A variable created inside a function (`local scope`) is like a patient's temporary vital signs used for a single calculation, whereas a universal constant like `LBS_TO_KG` (`global scope`) is like a widely accepted medical conversion factor available everywhere.
*   **`None` Type**: A special Python value representing the absence of a value or a null result. Its medical analogy is a lab test result coming back as "invalid sample" instead of a faulty number. It signals that a calculation could not be completed as intended, preventing the use of erroneous data.

### 3.1 Returning a Fixed Value from a Function

The simplest function encapsulates a single piece of logic or a value. We use the `return` keyword to output this value, allowing the result to be stored in a variable for later use. This ensures consistency across an application.

````python
def get_fever_threshold_c() -> float:
    """Returns a common clinical fever threshold in Celsius.

    Source: CDC. Fever and Taking Your Child's Temperature. 2021.

    Returns:
        float: The fever threshold value (38.0).
    """
    return 38.0

# The returned value is stored in the 'fever_threshold' variable.
fever_threshold = get_fever_threshold_c()
print(f"The fever threshold is {fever_threshold}°C.")
# Output:
# The fever threshold is 38.0°C.
````

### 3.2 Defining a Function with Parameters

Parameters make functions flexible, allowing them to operate on different input data for various patients or scenarios.

````python
# A constant accessible throughout the file.
LBS_TO_KG = 0.453592

def convert_lbs_to_kg(weight_lbs: float) -> float:
    """Converts a weight from pounds to kilograms.

    Args:
        weight_lbs (float): The weight in pounds.

    Returns:
        float: The converted weight in kilograms, rounded to 2 decimal places.
    """
    weight_kg = weight_lbs * LBS_TO_KG
    return round(weight_kg, 2)

# Here, 154.0 is the 'argument' passed to the 'weight_lbs' parameter.
patient_weight_kg = convert_lbs_to_kg(154.0)
print(f"Patient weight recorded: {patient_weight_kg} kg")
# Output:
# Patient weight recorded: 69.85 kg
````

### 3.4 Handling Invalid Inputs with `None`

Returning `None` from a function is a robust way to signal a calculation failure without crashing the program, which is essential for building safe clinical tools.

````python
def calculate_bmi(weight_kg: float, height_m: float) -> float | None:
    """Calculates Body Mass Index (BMI).

    Returns None if height is non-positive, as this is a physiological impossibility.
    Formula source: CDC. About Adult BMI. 2022.

    Args:
        weight_kg (float): Patient's weight in kilograms.
        height_m (float): Patient's height in meters.

    Returns:
        float | None: The calculated BMI rounded to one decimal place, or None if input is invalid.
    """
    if height_m <= 0:
        return None  # Signal invalid input.
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 1)

# Check if the function returned a valid number before using it.
patient_bmi = calculate_bmi(85, 1.75)
if patient_bmi is not None:
    print(f"Patient's BMI is: {patient_bmi}")
else:
    print("Could not calculate BMI due to invalid height.")
# Output:
# Patient's BMI is: 27.8
````

def calculate_pediatric_dose(weight_kg: float, dose_mg_per_kg: float, max_dose_mg: float = 500.0) -> Optional[float]:
    """Calculates a pediatric dose, enforcing a maximum dose limit.

    Args:
        weight_kg (float): The patient's weight in kilograms.
        dose_mg_per_kg (float): The prescribed dose rate in mg per kg.
