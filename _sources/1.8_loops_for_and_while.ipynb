## 1. Introduction

Welcome to the study of loops, the fundamental tools for automating repetitive tasks in programming. In precision health, data is abundant, and loops are essential for processing it efficiently. Whether you are analyzing a list of patient vitals, simulating drug decay over time, or monitoring a physiological signal until a target is met, loops are the engine that drives the analysis.

In this section, you will learn how to use Python's two primary loop structures: the `for` loop, which is perfect for iterating over a known sequence of items, and the `while` loop, which continues as long as a specific condition remains true. We will explore how to control loop behavior with `break` and `continue` and see how these constructs are applied in realistic clinical scenarios.

---

## 2. Key Concepts and Definitions

*   **Loop**: A control flow structure that allows a block of code to be executed repeatedly. In medicine, this is used for tasks like processing every patient record in a dataset or simulating a biological process over multiple time steps.
*   **Iteration**: A single execution of the code block inside a loop. For example, one iteration might involve checking the vital signs of a single patient from a list of many.
*   **Sequence**: An ordered collection of items that a `for` loop can iterate over. Common sequences include lists (e.g., a list of heart rate measurements) or strings (e.g., a DNA sequence).
*   **Condition**: A logical expression that evaluates to `True` or `False`, used to control a `while` loop. A medical analogy is titrating a drug drip: the condition is "Is the patient's blood pressure still above the target?" The loop continues as long as the answer is `True`.
*   **Infinite Loop**: A loop whose condition never becomes `False`, causing the program to run indefinitely and become unresponsive. This is a critical bug, analogous to a medical device that fails to shut off after a procedure is complete.
*   **`break` Statement**: A command that immediately terminates the innermost loop it is in. It's used to stop a process once a specific goal is met, such as halting a search for a genetic marker as soon as the first one is found.
*   **`continue` Statement**: A command that skips the remainder of the current iteration and proceeds to the next one. This is useful for ignoring invalid data, like skipping a negative heart rate reading in a list of vitals.

### 3.1 The `for` Loop

A `for` loop executes a block of code for each item in a sequence, making it ideal for processing lists of data.

```python
# Process a list of patient heart rates for tachycardia (>100 bpm)
heart_rates = [72, 105, 68, 95, 110]
for rate in heart_rates:
    # Tachycardia threshold for resting adults (AHA 2023)
    if rate > 100:
        print(f"Tachycardia detected: {rate} bpm")

# Output:
# Tachycardia detected: 105 bpm
# Tachycardia detected: 110 bpm
```

### 3.2 Nested `for` Loops

A nested loop is a loop inside another loop. This is useful for processing items within a collection, such as analyzing multiple vital signs for multiple patients.

```python

    "Patient B": [72, 145, 90]
}

# Define constants for vital sign indices for clarity
# Vitals are [Heart Rate (HR), Systolic BP (SBP), Diastolic BP (DBP)].

for patient, vitals in patients_vitals.items():
    print(f"Processing {patient}:")
    # Access SBP by its defined index
    sbp = vitals[SBP_INDEX]
    if sbp > 140: # Check for high SBP (>140 mmHg)
        print(f"  - ALERT: High SBP detected: {sbp} mmHg")

# Output:
```

### 3.3 The `while` Loop

A `while` loop repeats as long as a condition is `True`. This is perfect for situations where the number of iterations is unknown, such as titrating medication to a target physiological state.

```python
# Titrate medication until systolic BP is <= 130 mmHg
sbp = 160 # mmHg

# Target SBP (Whelton et al. 2018)
while sbp > 130:
    sbp -= 5
    print(f"BP after titration: {sbp} mmHg.")

print(f"Target reached. Final Systolic BP: {sbp} mmHg")

# Output:
# BP after titration: 155 mmHg.
# BP after titration: 150 mmHg.
# BP after titration: 145 mmHg.
# BP after titration: 140 mmHg.
# BP after titration: 135 mmHg.
# BP after titration: 130 mmHg.
# Target reached. Final Systolic BP: 130 mmHg
```

### 3.4 Loop Control: `break` and `continue`

Use `break` to exit a loop entirely and `continue` to skip the current iteration and move to the next.

The `break` statement is efficient for stopping a scan as soon as a target is found.

```python
# Scan patient temperatures for the first sign of fever (>38.0°C)
temperatures_c = [36.8, 38.3, 37.1, 38.2, 39.2, 38.6, 38.7]
for temp in temperatures_c:
    # Common fever threshold (CDC 2021)
    if temp > 38.0:
        print(f"Fever detected at {temp}°C. Halting scan.")
        break

# Output:
# Fever detected at 38.3°C. Halting scan.
```

The `continue` statement is ideal for skipping invalid or irrelevant data points.

```python
# Process only valid heart rate readings, ignoring errors (<= 0)
rates = [88, 95, -1, 102, 0, 99]
print("Processing valid heart rates:")
for rate in rates:
    if rate <= 0:
        print(f"  - Invalid reading ({rate}) skipped.")
        continue # Skips to the next rate
    print(f"  - Valid rate processed: {rate} bpm")

# Output:
# Processing valid heart rates:
#   - Valid rate processed: 88 bpm
#   - Valid rate processed: 95 bpm
#   - Invalid reading (-1) skipped.
#   - Valid rate processed: 102 bpm
```

### 3.5 The Loop `else` Clause

A loop's `else` block runs only if the loop completes all its iterations without encountering a `break` statement. This is useful for confirming that a search was completed without finding the target item.

```python
# Search for a critical potassium value, report if not found
lab_results = [4.1, 4.5, 5.0, 4.7] # Potassium levels in mEq/L
for result in lab_results:
    # Alert threshold for hyperkalemia (Simon et al. 2022)
    if result > 5.5:
        print("Critical value found. Alerting staff.")
        break
else:
    print("Scan complete. No critical values found.")

# Output:
# Scan complete. No critical values found.
```

## 5. Practical Applications

*   **Pharmacokinetic (PK) Modeling:** A `for` loop with `range()` can simulate how a drug's concentration changes in the body over several hours or days. Each iteration calculates the remaining drug based on its half-life, helping clinicians determine optimal dosing schedules.
*   **Real-Time Patient Monitoring:** A `while` loop can continuously process data from a bedside monitor (e.g., ECG, SpO2). The loop runs indefinitely, checking each new data point for critical abnormalities. A `break` statement might be triggered by a user action to stop monitoring, or an alert function could be called within the loop.
*   **Genomic Sequence Analysis:** `for` loops are fundamental for bioinformatics. A loop can iterate through a DNA sequence (a very long string) to count nucleotide frequencies, search for specific gene markers (e.g., 'ATG'), or identify patterns related to disease risk.
*   **Clinical Trial Data Auditing:** Nested `for` loops can be used to audit complex datasets. The outer loop might iterate through a list of trial participants, while an inner loop iterates through their scheduled visits, checking for protocol compliance, missing data, or adverse events.
