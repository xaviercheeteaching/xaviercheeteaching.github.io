## 1. Introduction

In precision health, data integrity is paramount. A program that crashes when encountering unexpected data—like a missing lab value or a disconnected medical device—can compromise patient care and research outcomes. This section introduces Python's error and exception handling capabilities, which are essential tools for building robust, fault-tolerant applications.

> **In Practice:** In a hospital's electronic health record (EHR) system, data comes from many sources: lab results, manual nurse entries, and device feeds. This data is often imperfect—containing text like "N/A" or being temporarily unavailable. Robust error handling isn't just a coding best practice; it's a patient safety requirement. It ensures the system remains stable and avoids crashing during critical moments, like when a doctor is accessing patient data during an emergency.

We will learn how to anticipate and manage runtime errors gracefully using `try...except`, ensure critical cleanup actions always run with `finally`, and signal our own errors with `raise`. By the end, you'll be able to write code that doesn't just work in ideal conditions but can also handle the messy, unpredictable reality of real-world data.

## 2. Key Concepts and Definitions

*   **Exception**: An error event that occurs during the execution of a program, disrupting its normal flow. In a medical context, an exception is like an unexpected vital sign reading that triggers a specific protocol instead of continuing with routine monitoring.
*   **`try...except` block**: A fundamental control structure for managing exceptions. The code inside the `try` block is monitored for errors. If an exception occurs, the code in the matching `except` block is executed. This is analogous to trying a medical procedure (`try`) and having a contingency plan (`except`) ready for a known complication.
*   **`finally` block**: A block of code that is always executed after the `try` and `except` blocks, regardless of whether an exception was raised or handled. This is like the mandatory sterilization of equipment after a surgical procedure—it must happen every single time, whether the operation was smooth or had complications.
*   **`raise` statement**: A command used to trigger an exception intentionally. This is useful for enforcing rules in your code. For example, you can `raise` an error if a function receives a patient age of -5, signaling that the input data is invalid and requires intervention.
*   **`ValueError`**: An exception raised when a function receives an argument of the correct data type but an inappropriate value. For instance, attempting to convert the text `"N/A"` into a number is a `ValueError` because the type (string) is valid for the function, but its content is not.
*   **`TypeError`**: An exception raised when an operation or function is applied to an object of an inappropriate type. A common example is trying to perform arithmetic between a number and a string, such as `500 + "mg"`, which is nonsensical in Python.
*   **`SyntaxError`**: An error indicating that the code breaks Python's grammar rules. This is like a grammatically incorrect sentence that cannot be understood. It must be fixed before the program can run.
*   **`NameError`**: An exception raised when you try to use a variable or function name that has not been defined. This is often caused by a simple typo.

### 3.1 Handling Runtime Errors with `try...except`

Use a `try...except` block to handle potential runtime errors without crashing the program. This is crucial when dealing with unpredictable data, such as converting patient data from various sources that may not be properly formatted.

````python
def calculate_risk_score(value: str, total_patients: int) -> None:
    """Calculates and prints a risk score, handling potential errors."""
    try:
        score = int(value) / total_patients
    except (ValueError, TypeError) as e:
        print(f"Data Error: Could not calculate score. Reason: {e}")
    except ZeroDivisionError:
        print("Data Error: Cannot divide by zero patients.")
    else:
        print(f"Calculated risk score: {score:.2f}")

# Call the function with different inputs
calculate_risk_score("95", 10)
calculate_risk_score("N/A", 10)
calculate_risk_score("95", None)
calculate_risk_score("95", 0)
````

Always catch the most specific errors you anticipate, like `ValueError` or `ZeroDivisionError`, so you can handle each one appropriately.

### 3.2 Guaranteed Cleanup with `finally`

Use `finally` to guarantee the execution of cleanup code, like closing a device connection, whether an error occurs or not.

````python
def read_monitor_data() -> None:
    """Simulates reading from a device and ensures cleanup."""
    is_connected = False
    try:
        print("Connecting to heart rate monitor...")
        is_connected = True
        # Simulate a read error from the monitor.
        raise RuntimeError("Monitor signal lost.")
    except RuntimeError as e:
        print(f"Error: {e}")
    finally:
        if is_connected:
            is_connected = False
            print("Monitor connection closed.")

read_monitor_data()
````

### 3.3 Signaling Errors with `raise`

Use `raise` to signal an error when data violates a rule, such as an invalid patient ID format.

````python
def process_patient(patient_id: str) -> None:
    """Processes a patient record, validating the ID format first."""
    if not (isinstance(patient_id, str) and patient_id.startswith("PT") and len(patient_id) == 8):
        raise ValueError("Patient ID must be a string in the format 'PT' followed by 6 digits.")
    print(f"Processing patient: {patient_id}")

try:
    process_patient("PT123456")   # Success case
    process_patient("P-12345")    # Failure case
except ValueError as e:
    print(f"Validation Error: {e}")
````

```python
    """Converts a string to a float and validates its clinical range."""
    try:

    except ValueError:
```

```python
def process_readings(readings_list: list[str]) -> None:
    """Processes a list of blood sugar readings and reports on data quality."""
    successful_count = 0
    failed_count = 0
    total_sum = 0
    
    try:
        print("Starting batch processing...")
        for reading in readings_list:
            try:
                value = int(reading)
                total_sum += value
                successful_count += 1
            except ValueError:
                failed_count += 1
        
        if successful_count > 0:
            average = total_sum / successful_count
            print(f"Average of valid readings: {average:.2f}")
        else:
            print("No valid readings to average.")
            
    finally:
        print("\n--- Processing Report ---")
        print(f"Successfully processed: {successful_count} readings")
        print(f"Failed to process: {failed_count} readings")
        print("-------------------------")

# Example Usage
patient_data = ["105", "98", "invalid", "110", "not recorded", "102"]
process_readings(patient_data)
```
```
# Expected Output:

# Average of valid readings: 103.75
# 
# --- Processing Report ---
# Successfully processed: 4 readings
# Failed to process: 2 readings
# -------------------------
```

**Explanation:** The outer `try...finally` ensures the report is always printed. The inner `try...except` handles conversion errors for each individual reading, allowing the loop to continue processing the rest of the list.
**Key Learning:** Combining nested `try...except` for item-level error handling with an outer `finally` for guaranteed summary actions.

## 5. Practical Applications

*   **Automated Lab Result Parsing**: In clinical data pipelines, scripts automatically parse files (like CSVs) containing lab results. These files often have inconsistencies, such as `"N/A"` instead of numeric values. Using a `try...except ValueError` block allows the script to skip malformed entries, log them for manual review, and continue processing the rest of the file without crashing, ensuring high data throughput.
*   **Medical Device Integration**: Software that communicates with medical devices (e.g., continuous glucose monitors, ECGs) must be resilient to connection failures. A `try...except RuntimeError` can handle sudden signal loss, while a `finally` block guarantees that the connection port is always closed properly. This prevents resource leaks and ensures the system remains stable and ready to reconnect.
*   **Genomic Data Validation**: Before running complex and costly genomic analyses, it's crucial to validate the input data. A function processing a DNA sequence can use `raise ValueError` if the sequence contains invalid characters (not A, T, C, or G). This "fail-fast" approach prevents corrupted data from propagating downstream and invalidating expensive research results.
