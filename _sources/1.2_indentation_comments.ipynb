## 1. Introduction

In this section, we will cover one of the most fundamental and unique features of Python: indentation. You will learn how to use indentation to define and structure code blocks. Unlike in many other programming languages where indentation is for readability, in Python, it is a core part of the syntax. This distinction is critical in a precision health setting, where a single indentation error could cause a vital safety check in a clinical algorithm to fail, potentially leading to incorrect diagnoses or patient harm. Mastering indentation is a foundational step toward writing safe, reliable, and effective medical software.

---

## 2. Key Concepts and Definitions

*   **Indentation**: The spaces at the beginning of a code line. In Python, indentation is a strict rule used to define the scope and structure of code. A medical analogy is the formatting of a prescription: the specific indentation of dosage lines separates them from patient instructions, and incorrect formatting can lead to dangerous misinterpretations.
*   **Code Block**: A group of one or more consecutive lines of code, all indented at the same level, that are executed as a single unit. In a clinical script, a code block might contain all the steps for an alert (e.g., notify nurse, log event, suggest action), which must only run together when a specific condition is met.
*   **IndentationError**: A specific error Python raises when code is not indented correctly. This error acts as a crucial safety mechanism, preventing logically flawed or ambiguous code from executing. In a medical context, this is like a pharmacist refusing to fill a prescription with an ambiguous dosage, preventing a potential medication error.

---

## 3. Main Content

In Python, control statements like `if`, `for`, and `while` require an indented block to group the code they affect.

> **Important:** In Python, indentation is syntax, not style. A single indentation error in a medical application could cause a critical safety check to fail, potentially leading to patient harm.

> **Pro Tip:** The universal standard for Python indentation is four spaces per level. While some code editors might use tabs, configuring your editor to always convert tabs to four spaces is a best practice that prevents collaboration errors and ensures your code is portable.

> **Debug Note:** If your indentation is incorrect, Python will raise an `IndentationError`. This is one of the most common errors for new Python programmers. Always check that each line in a code block is indented to the exact same level.

Let's look at an example. A script needs to check if a patient's temperature requires intervention.

```python
# Check if patient temperature requires intervention
patient_temp_c = 38.5
FEVER_THRESHOLD_C = 38.0 # Clinical thresholds are defined as constants

if patient_temp_c > FEVER_THRESHOLD_C:
    # This indented block executes only if the 'if' condition is true.
    print("Alert: Patient temperature is elevated, indicating fever.")
    print("Recommend antipyretic medication.")

# This line is not indented, so it is outside the block and always executes.
print("Temperature check complete.")

# Output:
# Alert: Patient temperature is elevated, indicating fever.
# Recommend antipyretic medication.
# Temperature check complete.
```
The two `print` statements under the `if` line are indented, creating a code block that executes only if `patient_temp_c` is greater than `FEVER_THRESHOLD_C`. The final `print` statement is not indented, so it is outside the block and always executes.

## 5. Practical Applications

*   **Clinical Decision Support (CDS):** In a hospital's electronic health record (EHR), Python scripts use `if` blocks to analyze incoming patient data. Correct indentation ensures that a multi-condition check—for fever, high white blood cell count, and low blood pressure—executes as a single unit to accurately trigger a sepsis alert. An indentation error could mean a life-threatening condition is missed.
*   **Automated Lab Reporting:** A script designed to flag abnormal lab results uses `if-elif-else` blocks to compare values against reference ranges. Proper indentation ensures that only the code for flagging a high potassium level runs when that specific condition is met, preventing the system from sending incorrect or ambiguous alerts to clinicians.
*   **Genomic Analysis Pipelines:** When processing DNA sequencing data, Python scripts use `for` loops and `if` statements to identify genetic variants. Indentation correctly groups the logic for annotating a specific gene mutation, ensuring that the right information is associated with the right variant before being added to a patient's genomic profile for personalized treatment.
