## 2. Key Concepts and Definitions

Understanding the following terms is essential for working with functions. They are the vocabulary we use to describe how functions are built and used.

*   **Function**: A named, reusable block of code designed to perform a single, specific task. In a hospital, think of it as a standardized protocol, like the steps for a patient check-in, which are performed the same way for every patient.
*   **`def` Keyword**: The Python keyword used to start a function definition. It signals to the Python interpreter that you are about to create a new function.
*   **Parameter**: A named placeholder in a function's definition that acts like a variable within the function. It specifies what kind of data the function expects to receive. For example, a `calculate_bmi` function would have `weight_kg` and `height_m` as parameters.
*   **Argument**: The actual value that is passed to a function when it is called. If a function is defined with a `patient_id` parameter, the string `"PT123456"` would be the argument you provide during the call.
*   **Docstring**: A string literal that appears as the first statement in a function's definition. It's used to document what the function does, what parameters it takes, and what it might return. It is a critical part of writing maintainable code.
*   **Type Hint**: An annotation that specifies the expected data type of a variable, function parameter, or return value (e.g., `patient_id: str`). They improve code clarity and help prevent bugs.
*   **Positional Argument**: An argument that is matched to a parameter based on its order (position) in the function call. The first argument maps to the first parameter, the second to the second, and so on.
*   **Keyword Argument**: An argument that is identified by the parameter name it corresponds to (`name=value`). This makes the function call more explicit and means the order of arguments no longer matters.

### 3.1 Defining and Calling a Basic Function

To create a function, you use the `def` keyword, followed by the function's name, parentheses `()`, and a colon `:`. Inside the parentheses, you define the function's **parameters** with type hints. The code belonging to the function is indented underneath. The first item inside the function should be a multi-line **docstring** explaining its purpose.

````python
def log_patient_entry(patient_id: str) -> None:
    """
    Prints a formatted check-in message for a given patient ID.

    Args:
        patient_id (str): The unique identifier for the patient.

    Returns:
        None: This function does not return any value; it prints to the console.
    """
    print(f"Patient {patient_id} has checked in.")

# '"PT123456"' is the ARGUMENT (the actual value passed during the call)
log_patient_entry("PT123456")
# Output:
# Patient PT123456 has checked in.
````

### 3.2 Returning Values and Using Arguments

Functions become even more powerful when they `return` a value that can be used elsewhere. This separates the calculation logic from the display logic. You can pass arguments to functions in two ways: by position or by keyword.

````python
def calculate_bmi(weight_kg: float, height_m: float) -> float:
    """
    Calculates Body Mass Index (BMI).

    Includes a safety check to prevent division by zero if height is not positive.

    Args:
        weight_kg (float): Patient's weight in kilograms.
        height_m (float): Patient's height in meters.

    Returns:
        float: The calculated BMI value, or 0.0 if height is non-positive.
    """
    if height_m <= 0:
        return 0.0  # Return a safe value to avoid errors
    bmi = weight_kg / (height_m ** 2)
    return bmi

# 1. Call with positional arguments and print the returned value
patient_bmi = calculate_bmi(70, 1.75)
print(f"Patient BMI: {patient_bmi:.2f}")
# Output:
# Patient BMI: 22.86

# 2. Call with keyword arguments (order is flexible)
patient_bmi = calculate_bmi(height_m=1.75, weight_kg=70)
print(f"Patient BMI: {patient_bmi:.2f}")
# Output:
# Patient BMI: 22.86
````

### 3.3 Argument Order Rules

When mixing argument types, there is one critical rule you must follow.

````python
# VALID: Positional argument comes first
patient_bmi = calculate_bmi(70, height_m=1.75)
print(f"Patient BMI: {patient_bmi:.2f}")
# Output:
# Patient BMI: 22.86

# INVALID: Positional argument cannot follow a keyword argument
# calculate_bmi(weight_kg=70, 1.75)  # -> SyntaxError
````

> **Debug Note:** The error `SyntaxError: positional argument follows keyword argument` is Python's way of telling you that you've placed a value without a label (a positional argument) after a value that has one (a keyword argument). Always place your "unlabeled" arguments first.

## 5. Practical Applications

Functions are the foundation of almost all real-world health data applications. Here are a few examples.

*   **Automating Patient Data Entry:** In hospital information systems, a function like `register_new_patient(name: str, dob: str, insurance_id: str)` can be used to create new patient records. This ensures that every new registration captures the same essential information in a standardized format, reducing data entry errors and improving data quality across the entire electronic health record (EHR) system.
*   **Calculating Clinical Risk Scores:** Many clinical decisions rely on standardized risk scores (e.g., CHADS₂ for stroke risk, APACHE II for ICU mortality risk). A function can be written for each score, such as `calculate_apache_score(age: int, temperature: float, lab_values: dict)`. This allows any clinician or analyst to compute the score accurately and consistently without needing to re-implement the complex scoring logic each time.
*   **Standardizing Genomic Data Pre-processing:** In bioinformatics, raw DNA sequencing data must undergo several cleaning and filtering steps before analysis. A function like `preprocess_fastq(file_path: str)` can encapsulate a standard pipeline (e.g., quality trimming, adapter removal). This ensures that all samples in a research study are processed identically, which is critical for the validity and reproducibility of the scientific findings.
