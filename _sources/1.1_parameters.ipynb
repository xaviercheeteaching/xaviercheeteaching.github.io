## 2. Key Concepts and Definitions

Understanding these terms is crucial for effectively using default parameters in your functions.

*   **Default Parameter**: A parameter in a function signature that is assigned a value using the `=` operator. If a function is called without providing an argument for this parameter, the default value is used automatically. For example, in a function `log_vitals(hr, bp, temp_unit='C')`, `temp_unit` is a default parameter.
*   **Required Parameter**: A parameter that does not have a default value. An argument must be explicitly provided for every required parameter when the function is called, otherwise Python will raise a `TypeError`. In a dosing function, `patient_weight` would be a required parameter.
*   **Function Signature**: The first line of a function definition, which includes the `def` keyword, the function name, and its parameters in parentheses. The order of required and default parameters in the signature is critical.
*   **Mutable Default Argument**: A default parameter that uses a mutable (changeable) data type, such as a list or dictionary. This is a common but dangerous pitfall, as the default object is shared across all calls to the function, leading to unintended side effects. It's like having every patient's new allergy added to a single, shared master list instead of their own individual record.
*   **`SyntaxError`**: An error raised by the Python interpreter when it encounters code that violates the language's grammatical rules. A common cause is placing a required parameter after a default parameter in a function's signature.

### 3.1 Defining Functions with Default Parameters

Default parameters make arguments optional by assigning a standard value if one isn't provided. This is incredibly useful for establishing sensible defaults in clinical calculations.

> **In Practice:**
> Default parameters simplify function calls for common medical scenarios. For instance, a prescription function can default to a standard 50 mg/mL drug concentration, reducing the risk of data entry errors for routine cases while still allowing for overrides in special situations.

````python
def calculate_med_dose(weight_kg: float, concentration_mg_ml: int = 50) -> None:
    """Calculates dose in mL based on weight and drug concentration.

    Args:
        weight_kg (float): Patient's weight in kilograms.
        concentration_mg_ml (int): Drug concentration in mg/mL. Defaults to 50.
    """
    # NOTE: Hypothetical 10mg/kg regimen for educational purposes only.
    dose_ml = (weight_kg * 10) / concentration_mg_ml
    print(f"Administer {dose_ml:.2f} mL for a {weight_kg} kg patient.")

# Call using the default concentration (50 mg/mL)
calculate_med_dose(70)

# Override the default with a different concentration
calculate_med_dose(70, concentration_mg_ml=100)
````

### 3.2 The Parameter Order Rule

> **Important:** Required Parameters Must Come First
>
> A strict rule in Python is that all required (non-default) parameters must be listed *before* any default parameters in a function definition. This ordering is essential for the Python interpreter to correctly match arguments to parameters. Violating this rule will cause a `SyntaxError`.

````python
# CORRECT: Required parameter (weight_kg) is first.
def calculate_med_dose(weight_kg: float, concentration_mg_ml: int = 50) -> None:
    pass

# INCORRECT: This will raise a SyntaxError.
# def calculate_med_dose(concentration_mg_ml: int = 50, weight_kg: float) -> None:
#     pass
````

### 3.3 The Pitfall of Mutable Default Arguments

> **Important:** Avoid Mutable Objects (like lists or dictionaries) as Default Arguments
>
> You must avoid using mutable objects like lists or dictionaries as default arguments. The default object is created only *once* when the function is defined, not each time it's called. This means every function call that relies on the default will modify the *same* object, leading to unexpected and incorrect behavior.
>
> The correct and safe pattern is to set the default value to `None` (which is immutable) and then, inside the function, check if the argument is `None`. If it is, you create a new, empty object (e.g., `[]` or `{}`). This ensures that each function call operates on a fresh, independent object.

````python
from typing import Optional, List

# INCORRECT: Using a mutable list as a default
def log_medication(medication: str, med_list: List[str] = []) -> None:
    med_list.append(medication)
    print(f"Medications logged: {med_list}")

log_medication("Lisinopril")
log_medication("Metformin") # This incorrectly modifies the original list
````

````python
# CORRECT: Use None and create the list inside the function
def log_medication_correct(medication: str, med_list: Optional[List[str]] = None) -> None:
    if med_list is None:
        med_list = []
    med_list.append(medication)
    print(f"Medications logged: {med_list}")

log_medication_correct("Lisinopril")
log_medication_correct("Metformin") # This correctly creates a new list
````

```python
def check_vitals(resting_heart_rate: int, patient_status: str = 'stable') -> None:
    """Checks patient vitals and flags potential tachycardia.

    Args:
        resting_heart_rate (int): The patient's resting heart rate (HR) in beats per minute.
        patient_status (str): The patient's current status. Defaults to 'stable'.
    """
    if patient_status != 'stable':
        print(f'Vitals logged for {patient_status} patient')
    elif resting_heart_rate > TACHYCARDIA_THRESHOLD_BPM:
        print('Tachycardia detected in stable patient')
    else:
        print('Vitals normal for stable patient')

# Test cases
check_vitals(110)
check_vitals(95)
check_vitals(110, 'unstable')
```
```text
# Output:

Vitals normal for stable patient
Vitals logged for unstable patient
```
**Explanation:** The function first checks for the non-default case (`'unstable'`) to prioritize it. If the patient is stable, it then evaluates the heart rate against the `TACHYCARDIA_THRESHOLD_BPM` constant.
**Key Learning:** Combining default parameters with conditional logic allows for creating flexible functions that can adapt to different clinical scenarios.

## 5. Practical Applications

*   **Electronic Health Record (EHR) Systems:** When logging a patient encounter, a function like `log_visit(patient_id, diagnosis_code, visit_type='Routine')` can use a default parameter for `visit_type`. This speeds up data entry for the most common encounters while allowing for specific overrides like `'Specialist Consultation'` or `'Emergency'`.
*   **Clinical Decision Support:** A function to assess sepsis risk, `check_sepsis_risk(vitals, lactate, consciousness_level='Alert')`, could default the consciousness level to 'Alert'. This allows the function to be called with just vitals and lab values for most patients, but enables an override for patients with altered mental status, a key sepsis indicator.
*   **Genomic Sequencing Pipelines:** In bioinformatics, a function that aligns DNA sequences might have default parameters for alignment sensitivity. For example, `align_sequence(sequence, reference_genome, sensitivity='high')` could default to a high-sensitivity setting for clinical-grade analysis but allow bioinformaticians to set it to `'fast'` for preliminary quality checks, saving significant computation time.
