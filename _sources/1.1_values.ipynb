## 2. Key Concepts and Definitions

Understanding the following terms is essential for building functions that produce results. They provide the vocabulary for describing how data flows out of your reusable code blocks.

*   **`return` Statement**: A keyword in Python that immediately exits a function and optionally passes a value back to the caller. In a medical context, this is like a lab test that "returns" a result (e.g., a blood glucose level) to the clinician who ordered it.
*   **Return Value**: The specific data or object that a function sends back. This could be a single number (like a calculated BMI), a string (a risk assessment category), or a more complex data structure.
*   **Tuple**: An ordered, immutable collection of items in Python. Functions use tuples to return multiple values at once, packed into a single object. For example, a function could return a patient's systolic and diastolic pressures as `(120, 80)`.
*   **`None`**: A special singleton object in Python representing the absence of a value. In clinical programming, returning `None` is a standard and effective way to signal that a calculation could not be completed due to invalid input, such as a negative patient weight.
*   **Conditional Logic**: The use of statements like `if`, `elif`, and `else` to control the flow of a program. In functions, conditional logic is often used with `return` to exit early if certain conditions are met, preventing errors and improving efficiency.

### 3.1 Return a Single Value

A function can process data and send back one piece of information. This is the most common use of the `return` statement. The returned value can be stored in a variable for later use.

````python
def calculate_drug_dosage(patient_weight_kg: float, dose_mg_per_kg: float) -> float:
    """Calculates total dosage based on patient weight.

    Args:
        patient_weight_kg (float): The patient's weight in kilograms.
        dose_mg_per_kg (float): The required dose in mg per kg.

    Returns:
        float: The total calculated dosage in mg.
    """
    total_dosage = patient_weight_kg * dose_mg_per_kg
    return total_dosage

dosage_for_patient = calculate_drug_dosage(70, 1.5)
print(f"Calculated Dosage: {dosage_for_patient} mg")
# Output:
# Calculated Dosage: 105.0 mg
````

### 3.2 Return Multiple Values

To return multiple values, you simply list them after the `return` keyword, separated by commas. Python automatically packs these values into a single tuple object. This tuple can then be "unpacked" into multiple variables, which is highly efficient for returning related pieces of information.

````python
def get_sample_vitals() -> tuple[float, int]:
    """Returns a sample of a patient's core vital signs.

    Returns:
        tuple[float, int]: A tuple containing temperature in Celsius and pulse in bpm.
    """
    temperature_c = 37.0
    pulse_bpm = 72
    return temperature_c, pulse_bpm

# The returned values are packed into a single tuple
vitals_tuple = get_sample_vitals()
print(f"Returned tuple: {vitals_tuple}")
# Output:
# Returned tuple: (37.0, 72)

# The tuple can be "unpacked" into separate variables
temp, pulse = get_sample_vitals()
print(f"Unpacked - Temperature: {temp} °C, Pulse: {pulse} bpm")
# Output:
# Unpacked - Temperature: 37.0 °C, Pulse: 72 bpm
````

### 3.3 Early Return with Conditional Logic

You can place a `return` statement inside a conditional block (`if` statement) to exit a function early. This is a powerful pattern for input validation, as it stops the function from executing further if the inputs are invalid. Returning `None` is a common convention to signal a failure.

````python
def calculate_bmi(weight_kg: float, height_m: float) -> float | None:
    """Calculates Body Mass Index (BMI), returning None if inputs are invalid.

    Args:
        weight_kg (float): Patient's weight in kilograms.
        height_m (float): Patient's height in meters.

    Returns:
        float | None: The calculated BMI, or None if inputs are non-positive.
    """
    if height_m <= 0 or weight_kg <= 0:
        return None  # Exit early if data is invalid
    
    bmi = weight_kg / (height_m ** 2)
    return bmi

# Failure case: Invalid height
patient_bmi_invalid = calculate_bmi(70, 0)
if patient_bmi_invalid is None:
    print("Error: Height and weight must be positive numbers.")
else:
    print(f"Patient BMI: {patient_bmi_invalid:.2f}")
# Output:
# Error: Height and weight must be positive numbers.

# Success case: Valid inputs
patient_bmi_valid = calculate_bmi(70, 1.75)
if patient_bmi_valid is None:
    print("Error: Height and weight must be positive numbers.")
else:
    print(f"Patient BMI: {patient_bmi_valid:.2f}")
# Output:
# Patient BMI: 22.86
````

> **Important:**
> Using an early `return` to handle invalid inputs is a fundamental safety pattern in medical software, often called a "guard clause." It stops a calculation cold if the input data is nonsensical (like a negative height). This prevents a "garbage in, garbage out" scenario, ensuring that an erroneous calculation never reaches the clinician.

> **Debug Note:**
> A common error is forgetting to check if a function returned `None`. If you try to perform math on `None` (e.g., `print(f"{patient_bmi_invalid:.2f}")` without the `if` check), your program will crash with a `TypeError`. Always handle the possibility of a `None` return value.

### 3.4 Implicit `None` Return

A function that completes its execution without encountering an explicit `return` statement will automatically return `None`. This is common for functions designed only to perform an action (a "side effect"), such as printing to the console, and do not need to produce a result.

````python
def log_abnormal_vital(patient_id: str, message: str) -> None:
    """Logs a warning message to the console. Does not return a value.

    Args:
        patient_id (str): The patient's unique identifier.
        message (str): The warning message to log.
    """
    print(f"[ALERT] Patient {patient_id}: {message}")

# The function prints a message but doesn't explicitly return a value
result = log_abnormal_vital("PT001234", "Heart rate is 110 bpm")
print(f"Return value of log_abnormal_vital: {result}")
# Output:
# [ALERT] Patient PT001234: Heart rate is 110 bpm
# Return value of log_abnormal_vital: None
````

```python
def calculate_map(systolic: float, diastolic: float) -> float | None:
    """Calculates Mean Arterial Pressure (MAP).

    Args:
        systolic (float): Systolic blood pressure.
        diastolic (float): Diastolic blood pressure.

    Returns:
        float | None: The calculated MAP, or None if inputs are invalid.
    """
    if not systolic > diastolic:
        return None  # Invalid physiological state
    
    map_value = diastolic + (1/3 * (systolic - diastolic))
    return map_value

# Valid call
map_result = calculate_map(120, 80)
if map_result is not None:
    print(f"The patient's MAP is: {map_result:.2f} mmHg")

# Invalid call
map_result_invalid = calculate_map(80, 90)
if map_result_invalid is None:
    print("Calculation failed: Systolic pressure must be greater than diastolic pressure.")
```
**Explanation:** The function first implements a "guard clause" to check if systolic pressure is greater than diastolic. If not, it immediately returns `None`. Otherwise, it proceeds to calculate and return the MAP, which the calling code then formats to two decimal places.
**Key Learning:** Using an early `return` with `None` is an effective way to handle physiologically impossible or invalid data before performing calculations.

## 5. Practical Applications

*   **Clinical Risk Scoring:** A function can take multiple patient parameters (e.g., age, cholesterol, blood pressure) and `return` a calculated risk score, such as the Framingham Risk Score for cardiovascular disease. *   **Genomic Data Analysis:** In bioinformatics, a function might process a raw DNA sequence file. *   **Data Validation and Cleaning:** When processing electronic health records (EHR), a function can be written to validate patient data. If the data is valid, it might `return` a cleaned, standardized dictionary of information.
