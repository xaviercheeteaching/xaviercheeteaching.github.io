## 2. Key Concepts and Definitions

*   **Nested Conditional**: An `if`, `if-else`, or `if-elif-else` statement that is placed inside another conditional block. The inner condition is only evaluated if the outer condition is true, creating a dependent decision path.
*   **Decision Hierarchy**: A structured sequence of choices where later decisions depend on the outcomes of earlier ones. This is analogous to a clinical triage flowchart, where a patient's path is determined by a series of sequential checks.
*   **Code Block**: A group of Python statements indented at the same level. In nested conditionals, the inner block has a deeper level of indentation than the outer block, visually representing the decision hierarchy.
*   **Glasgow Coma Scale (GCS)**: A neurological scale used by clinicians to assess a person's level of consciousness after a brain injury. A lower score indicates a more severe impairment, while a score of 15 represents a fully conscious state.
*   **Tachycardia**: A medical term for a heart rate that is faster than the normal resting rate. In adults, this is often defined as over 100 beats per minute and can be a sign of stress, illness, or other conditions.

### 3.1 The Structure of Nested Conditionals

Nested conditionals are essential for creating logic that follows a sequence of dependent checks. The inner `if` statement is only executed if the outer `if` condition evaluates to `True`. This is controlled by indentation; the nested block is indented further to the right.

Consider a common clinical scenario: checking for fever and then, only if a fever is present, checking for a high heart rate (tachycardia), which could be a sign of sepsis.

````python
# Clinical thresholds (constants)
FEVER_THRESHOLD_C = 38.0
TACHYCARDIA_THRESHOLD_BPM = 100

# Patient vitals record stored in a dictionary
patient_vitals = {
    "temperature_c": 38.5,
    "heart_rate_bpm": 110
}

# Outer condition: check for fever
if patient_vitals["temperature_c"] > FEVER_THRESHOLD_C:
    print(f"ALERT: Fever detected (Temp > {FEVER_THRESHOLD_C}°C).")
    
    # Inner (nested) condition: check for tachycardia
    # This check only runs if the first condition (fever) is True.
    if patient_vitals["heart_rate_bpm"] > TACHYCARDIA_THRESHOLD_BPM:
        print(f"ALERT: Tachycardia detected (HR > {TACHYCARDIA_THRESHOLD_BPM} bpm). Assess for sepsis.")
    else:
        print("Heart rate is within normal limits.")
else:
    print("Patient temperature is normal.")
````

> **Important:** In Python, indentation is not just for readability—it defines the structure of your code. The indented code block inside an `if` statement only runs when that condition is `True`. For nested conditionals, you will have a second level of indentation. Incorrect indentation will cause errors or lead to unexpected behavior.

> **Pro Tip: Nesting vs. Logical Operators**
> If a nested check has no `else` branch, you can often simplify it with a logical `and` operator. For example, `if patient_vitals["temperature_c"] > FEVER_THRESHOLD_C and patient_vitals["heart_rate_bpm"] > TACHYCARDIA_THRESHOLD_BPM:` is sometimes clearer. Use nesting when you must handle intermediate steps separately, such as printing the "Fever detected" alert *before* checking the heart rate.

## 5. Practical Applications

*   **Automated Clinical Decision Support Systems (CDSS)**: In electronic health record (EHR) systems, nested logic is used to guide clinicians. For example, if a patient's lab results show low potassium (outer condition), the system might then check if they are prescribed a diuretic (inner condition) and flag a potential drug-induced electrolyte imbalance.
*   **Pharmacogenomics Alerting**: These systems use nested conditionals to personalize medicine. The first check is for a specific drug prescription (e.g., Warfarin). If the drug is present, a nested check queries the patient's genetic data for variants like `CYP2C9` or `VKORC1`. If a relevant variant is found, an alert is triggered to recommend a non-standard dose.
*   **Intelligent Ventilator Management**: In an ICU, a monitoring system can use nested logic to manage ventilated patients. The outer condition might check if a patient's oxygen saturation drops. If it does, a nested condition could check the current ventilator settings (e.g., `FiO2`) to see if they are already maxed out, triggering an alarm for a physician to intervene manually.
