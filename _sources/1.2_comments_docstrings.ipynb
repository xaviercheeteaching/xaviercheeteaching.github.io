## 2. Key Concepts and Definitions

*   **Comment (`#`)**: A note within the code intended for developers that is ignored by the Python interpreter. In a medical context, comments are crucial for explaining the clinical reasoning behind a specific calculation, citing a research paper for an algorithm (e.g., a specific GFR estimation formula), or warning about potential data pitfalls.
*   **Docstring (`"""..."""`)**: A string literal that occurs as the first statement in a module, function, class, or method definition. It is not ignored by Python; instead, it becomes the `__doc__` attribute of that object. Its purpose is to document what the code does for the *user* of the function. In healthcare IT, tools can automatically parse docstrings to generate technical manuals for clinical software.
*   **Code Readability**: A measure of how easily a human reader can comprehend the purpose, control flow, and operation of source code. High readability is paramount in regulated medical software, as it facilitates peer review, debugging, and validation by regulatory bodies, ensuring patient safety and software reliability.

### 3.2 Docstrings: Explaining the 'What'

A docstring is created using triple quotes (`"""..."""`) and must be the very first statement inside a function. It describes the function's purpose, arguments, and return value.

> **Pro Tip:** A simple way to remember the difference is:
> *   **Comments (`#`)** explain the **why** to the *next developer* who reads your code.
> *   **Docstrings (`"""..."""`)** explain the **what** to the *user* of your function, often through automated documentation.

The following example demonstrates a function with type hints, error handling, and a complete docstring.

````python
def calculate_drug_dosage(patient_weight_kg: float, concentration_mg_ml: float, dose_mg_per_kg: float) -> float:
    """Calculates the required volume of a drug to administer.

    Example:
        >>> calculate_drug_dosage(70, 50, 2)
        2.8

    Args:
        patient_weight_kg: The patient's weight in kilograms.
        concentration_mg_ml: The drug concentration in mg/mL.
        dose_mg_per_kg: The prescribed dosage in mg per kg.

    Returns:
        The volume of the drug to administer in milliliters (mL).

    Raises:
        ValueError: If concentration_mg_ml is not a positive number.
    """
    if concentration_mg_ml <= 0:
        raise ValueError("Drug concentration must be a positive value.")
    
    total_mg_needed = patient_weight_kg * dose_mg_per_kg
    volume_ml = total_mg_needed / concentration_mg_ml
    return round(volume_ml, 2)

# Example usage and output
administered_volume = calculate_drug_dosage(patient_weight_kg=70, concentration_mg_ml=50, dose_mg_per_kg=2)
print(f"Required drug volume: {administered_volume} mL")

# Output:
# Required drug volume: 2.8 mL
````

```python
# The logic for staging is based on the serum creatinine criteria from the
# KDIGO 2012 Clinical Practice Guideline for Acute Kidney Injury.
# This function simplifies the criteria and does not include urine output.
def get_aki_stage(creatinine_increase_factor: float) -> int:
    """Determines the AKI stage based on creatinine increase from baseline.

    Args:
        creatinine_increase_factor: The fold-increase in serum
            creatinine from the patient's baseline value.

    Returns:
        The AKI stage (0, 1, 2, or 3). Returns 0 if criteria for
             stage 1 are not met.
    """
    if creatinine_increase_factor >= 3.0:
        return 3  # Stage 3: 3.0-fold increase or more
    elif creatinine_increase_factor >= 2.0:
        return 2  # Stage 2: 2.0-2.9-fold increase
    elif creatinine_increase_factor >= 1.5:
        return 1  # Stage 1: 1.5-1.9-fold increase
    else:
        return 0  # No AKI by this criterion

# Example usage:
stage = get_aki_stage(creatinine_increase_factor=2.5)
print(f"Patient AKI Stage: {stage}")

# Output:
# Patient AKI Stage: 2
```

```python
def parse_vitals_string(vitals_str: str) -> dict:
    """Parses a string of vital signs into a dictionary.

    The string is expected to be a comma-separated list of key-value pairs,
    where each pair is joined by a colon.

    Example:
        >>> parse_vitals_string("HR:78,BP:120/80,SPO2:98")
        {'HR': '78', 'BP': '120/80', 'SPO2': '98'}
        >>> parse_vitals_string("")
        {}

    Args:
        vitals_str: A string containing the vital signs data.

    Returns:
        A dictionary where keys are the vital sign names and values
        are the string representations of their measurements.
    """
    if not vitals_str:
        return {}
    vitals_dict = {}
    pairs = vitals_str.split(',')
    for pair in pairs:
        parts = pair.split(':', 1)
        if len(parts) == 2:
            key, value = parts
            vitals_dict[key.strip()] = value.strip()
    return vitals_dict

# Example usage:
vitals_data = "HR:78,BP:120/80,SPO2:98"
print(f"Parsed Vitals: {parse_vitals_string(vitals_data)}")

# Output:
# Parsed Vitals: {'HR': '78', 'BP': '120/80', 'SPO2': '98'}
```
**Explanation:** The `Example` block contains doctests that can be automatically run to verify the function's correctness. This practice integrates testing directly into the documentation.
**Key Learning:** Doctests are a powerful way to provide clear examples that also serve as verifiable tests, improving code reliability.

## 5. Practical Applications

Docstrings written in a standard format can be processed by tools like Sphinx to automatically generate a searchable HTML documentation website. This allows a hospital's IT staff to quickly understand how to integrate the tool and clinicians to learn how to use it effectively.
