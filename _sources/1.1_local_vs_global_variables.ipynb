## 1. Introduction

Welcome to our exploration of variable scope in Python. In this section, we will learn how to manage variable access using local, global, and nonlocal scopes to prevent data conflicts. Understanding scope is not just a technical exercise; it's a critical skill for maintaining data integrity in high-stakes environments like healthcare. Our journey will cover the boundaries of local variables, how to intentionally modify global state, and the nuanced control offered by the `nonlocal` keyword.

> **In Practice:**
> Proper scope control is essential for building safe clinical software. It prevents functions from accidentally overwriting critical shared data, such as a patient's Electronic Health Record (EHR) or a global configuration setting. This ensures that programs are robust, predictable, and that a function processing one patient's data cannot inadvertently affect another's.

---

## 2. Key Concepts and Definitions

Understanding the following terms is essential for managing data flow and preventing errors in your programs. They define the rules for where a variable can be accessed.

*   **Scope**: The region of a program where a variable is directly accessible. In a clinical setting, this is like data access levels—a nurse might have access to a patient's local chart (local scope), while a hospital administrator has access to all records (global scope).
*   **Local Variable**: A variable declared inside a function. It is only accessible from within that function. Think of it as a temporary note written on a patient's bedside clipboard; it's relevant only in that specific context and is discarded after the doctor's visit.
*   **Global Variable**: A variable declared in the main body of a script. It is accessible throughout the entire program. This is analogous to a hospital-wide alert system status, which can be read and modified by authorized functions in different departments.
*   **Variable Shadowing**: When a local variable has the same name as a global variable, the local variable takes precedence within its function. This "shadows" the global variable, preventing the function from seeing or modifying it unintentionally, thereby protecting the global state.
*   **LEGB Rule**: The order in which Python searches for a variable: **L**ocal, **E**nclosing, **G**lobal, and finally **B**uilt-in. This is the exact protocol Python follows to resolve a variable's value, similar to how a hospital might first check the local ward, then the department, then central records to find a patient's file.
*   **Nonlocal Variable**: A variable that is not local to the current function but is also not global. It resides in the scope of an enclosing (parent) function. The `nonlocal` keyword is used to modify such variables, like a junior doctor updating a patient status that is being managed by a senior doctor in the same team.

### 3.1 Local Scope & Shadowing

A variable created inside a function has **local scope**, meaning it only exists within that function. If it shares a name with a global variable, the local version "shadows" the global one, protecting the global state from accidental changes. This is the default and safest behavior for variables in functions.

> **Key Terms:**
> *   **Scope:** The region of a program where a variable can be accessed.
> *   **Shadowing:** When a variable in a local scope has the same name as a variable in an outer (e.g., global) scope, temporarily hiding the outer variable.

````python
# No imports needed for this example

patient_status = "Stable" # Global variable

def log_local_event() -> None:
    """Demonstrates a local variable shadowing a global one."""
    # This assignment creates a new LOCAL variable, 'patient_status'.
    patient_status = "Under Observation"
    print(f"  Inside function: Local status is '{patient_status}'")

print(f"Before call: Global status is '{patient_status}'")
log_local_event()
print(f"After call: Global status is still '{patient_status}'")

# Output:
# Before call: Global status is 'Stable'
#   Inside function: Local status is 'Under Observation'
# After call: Global status is still 'Stable'
````

### 3.2 The `global` Keyword

To modify a global variable from within a function, you must explicitly declare your intent with the `global` keyword. This is like requesting clearance to update a patient's master file, making a permanent, system-wide change.

````python
# No imports needed for this example

on_call_provider = "Dr. Smith" # Global variable

def change_shift(new_provider: str) -> None:
    """Updates the global on-call provider."""
    global on_call_provider
    on_call_provider = new_provider
    print(f"  Shift Update: {on_call_provider} is now on call.")

print(f"Initial on-call provider: {on_call_provider}")
change_shift("Dr. Jones")
print(f"Current on-call provider: {on_call_provider}")

# Output:
# Initial on-call provider: Dr. Smith
#   Shift Update: Dr. Jones is now on call.
# Current on-call provider: Dr. Jones
````

### 3.3 Modifying Global Mutable Objects

You do not need the `global` keyword to modify the *contents* of a mutable object like a list or dictionary. The keyword is only required to reassign the entire variable to a new object.

````python
# No imports needed for this example

patient_meds = ['aspirin'] # Global list

def add_medication(new_med: str) -> None:
    """Appends a medication to the global list."""
    # This modifies the original list's contents. No 'global' keyword needed.
    patient_meds.append(new_med)

def replace_med_list(new_list: list) -> None:
    """Reassigns the global medication list to a new list."""
    global patient_meds # Required to reassign the global variable
    patient_meds = new_list

print(f"Initial meds: {patient_meds}")
add_medication('lisinopril')
print(f"After adding: {patient_meds}")
replace_med_list(['ibuprofen'])
print(f"After replacing: {patient_meds}")

# Output:
# Initial meds: ['aspirin']
# After adding: ['aspirin', 'lisinopril']
# After replacing: ['ibuprofen']
````

### 3.4 The LEGB Rule in Action

Python finds variables by searching scopes in this order: **L**ocal, **E**nclosing, **G**lobal, **B**uilt-in.

````python
# No imports needed for this example

hospital_name = "General Hospital" # G: Global scope

def assign_ward() -> None:
    """Demonstrates enclosing scope."""
    ward = "Cardiology" # E: Enclosing scope
    
    def assign_patient() -> None:
        """Demonstrates local scope and LEGB lookup."""
        patient_id = "PT000456" # L: Local scope
        # 'print' is found in the B: Built-in scope.
        # 'patient_id' is found in the L: Local scope.
        # 'ward' is found in the E: Enclosing scope.
        # 'hospital_name' is found in the G: Global scope.
        print(f"Assigning {patient_id} to {ward} at {hospital_name}.")

    assign_patient()

assign_ward()

# Output:
# Assigning PT000456 to Cardiology at General Hospital.
````

### 3.5 The `nonlocal` Keyword

The `nonlocal` keyword allows a nested function to modify a variable in its enclosing (but not global) scope. This is useful for creating functions with internal "memory" or state without polluting the global scope.

> **In Practice:**
> The `nonlocal` keyword is useful for creating helper functions that manage state. Imagine a main function processing a patient's lab results that uses a nested helper function to track the count of abnormal readings. `nonlocal` would allow the nested function to update the `abnormal_count` variable belonging to the main processing function.

````python
# No imports needed for this example

def manage_patient_alerts() -> None:
    """Demonstrates using 'nonlocal' to manage internal state."""
    alert_level = 'Green' # Enclosing scope variable

    def escalate() -> None:
        """A nested function that modifies the enclosing scope's variable."""
        nonlocal alert_level
        alert_level = 'Yellow'

    print(f"Initial alert level: {alert_level}")
    escalate()
    print(f"Final alert level: {alert_level}")

manage_patient_alerts()

# Output:
# Initial alert level: Green
# Final alert level: Yellow
````

def monitor_heart_rate() -> str:
    """
    Initializes a patient status and uses a nested function
    to check a new reading and update the status.
    """
    status = 'Normal' # Enclosing scope variable
    
    def check_reading(new_rate: int) -> None:

        if not (50 <= new_rate <= 120):

**Key Learning:** The `nonlocal` keyword is the correct tool for a nested function to alter the state of its enclosing function without affecting the global scope.

# No imports needed for this example
from typing import Callable

# Global state
total_entries = 0
log_file = []

def create_log_session(user: str) -> Callable[[str | None], None]:
    """Creates a logging session with its own entry counter."""
    session_entries = 0 # Enclosing scope variable
    
    def log_data(data_point: str | None) -> None:
        """Logs a data point, updating session and global counters."""
        # Keywords to modify variables in outer scopes
        global total_entries
        nonlocal session_entries
        
        is_valid = data_point is not None # Local scope variable
        
        if is_valid:
            # Increment counters
            total_entries += 1
            session_entries += 1
            
            # Modify global mutable list (no keyword needed for modification)
            log_message = f"[USER: {user}] {data_point}"
            log_file.append(log_message)
            
    return log_data

# --- DO NOT EDIT BELOW THIS LINE ---

**Key Learning:** Complex functions can require simultaneous management of all scope levels. Using `global` and `nonlocal` precisely allows for state management at different granularities.

## 5. Practical Applications

*   **Global Configuration Management:** In a large-scale data processing pipeline for genomic data, a global dictionary might hold configuration parameters like file paths or quality score thresholds. A function designed to update this configuration for a new run would use the `global` keyword to ensure all subsequent modules in the pipeline use the updated settings.
*   **Real-Time Patient Monitoring Dashboards:** A function managing a patient's dashboard might use nested functions to handle different data streams (e.g., ECG, blood pressure). The parent function could hold a patient `status` variable (e.g., 'Stable', 'Watch', 'Critical'). A nested function processing ECG data could use the `nonlocal` keyword to escalate this `status` if it detects an arrhythmia, immediately updating the state managed by the parent function.
*   **Isolated Data Validation Modules:** When processing incoming patient data from an external lab, you can write a validation function that takes a record as an argument. Inside this function, all temporary variables used for checks (e.g., `is_date_valid`, `is_value_in_range`) are in the **local scope**. This guarantees that the validation logic is self-contained and cannot accidentally modify the original patient record or any global variables.
