    """Validates a lab result and returns (value, status)"""
    try:
        value = float(value_str)
        if 10 <= value <= 20:
            status = "normal"
        else:
            status = "abnormal"
        return (value, status)
    except ValueError:
        return (None, "invalid")

# Test the function

    else:
        print(f"  {test} -> {status}")

print(f"Average Systolic: {avg_systolic:.1f} mmHg")

    """Check for dangerous drug interactions"""
    interactions_found = []
    
    # Check all pairs of medications
    for i in range(len(prescribed_meds)):

def analyze_trial_data(patient_data):
    """Analyze clinical trial results"""
    
    improved = 0
    unchanged = 0
    worsened = 0
    dropouts = 0
    total_change = 0
    completers = 0
    
    results = []
    
    for patient in patient_data:
        patient_id = patient["id"]
        baseline = patient["baseline"]
        followup = patient.get("followup")
        
        # Handle dropouts
        if followup is None:
            dropouts += 1
            results.append({
                "id": patient_id,
                "status": "Dropped out",
                "change": None
            })
            continue
        
        # Calculate change (negative means improvement for things like BP, glucose)
        change = followup - baseline
        completers += 1
        total_change += change
        
        # Categorize outcome
        if change <= -10:
            status = "Improved"
            improved += 1
        elif change < 0 or abs(change) < 10:
            status = "Unchanged"
            unchanged += 1
        else:
            status = "Worsened"
            worsened += 1
        
        results.append({
            "id": patient_id,
            "baseline": baseline,
            "followup": followup,
            "change": change,
            "status": status
        })
    
    # Calculate statistics
    total_patients = len(patient_data)
    dropout_rate = (dropouts / total_patients) * 100
    avg_change = total_change / completers if completers > 0 else 0
    
    return {
        "results": results,
        "improved": improved,
        "unchanged": unchanged,
        "worsened": worsened,
        "dropouts": dropouts,
        "dropout_rate": dropout_rate,
        "average_change": avg_change

def analyze_lab_time_series(measurements):
    """Comprehensive analysis of lab values over time"""
    
    valid_readings = []
    missing_days = []
    rapid_changes = []
    
    # Process and validate data
    for day, value in measurements:
        try:
            numeric_value = float(value)
            valid_readings.append((day, numeric_value))
        except (ValueError, TypeError):
            missing_days.append(day)
    
    # Calculate basic statistics
    if len(valid_readings) == 0:
        return "Error: No valid readings found"
    
    values_only = [val for day, val in valid_readings]
    average = sum(values_only) / len(values_only)
    
    # Determine trend
    first_value = values_only[0]
    last_value = values_only[-1]
    change_percent = ((last_value - first_value) / first_value) * 100
    
    if change_percent > 10:
        trend = "Increasing"
    elif change_percent < -10:
        trend = "Decreasing"
    else:
        trend = "Stable"
    
    # Check for rapid changes between consecutive readings
    for i in range(1, len(valid_readings)):
        prev_day, prev_val = valid_readings[i-1]
        curr_day, curr_val = valid_readings[i]
        
        percent_change = abs((curr_val - prev_val) / prev_val) * 100
        
        if percent_change > 15:
            rapid_changes.append({
                "from_day": prev_day,
                "to_day": curr_day,
                "change": curr_val - prev_val,
                "percent": percent_change
            })
    
    return {
        "valid_readings": len(valid_readings),
        "average": average,
        "trend": trend,
        "change_percent": change_percent,
        "missing_days": missing_days,
        "rapid_changes": rapid_changes,
        "readings": valid_readings
    }

def assign_triage_priority(patient, critical_symptoms):
    """Assign triage priority based on symptoms and vitals"""
    
    patient_id = patient.get("id", "Unknown")
    symptoms = patient.get("symptoms", set())
    vitals = patient.get("vitals", {})
    
    # Extract vitals with defaults
    temp = vitals.get("temp", 37.0)
    hr = vitals.get("hr", 70)
    bp_sys = vitals.get("bp_sys", 120)
    
    # Check for critical symptoms
    has_critical = len(symptoms & critical_symptoms) > 0
    
    # Determine priority
    if has_critical or (temp > 39.0 and hr > 120):
        priority = 1
        category = "CRITICAL"
        reason = []
        if has_critical:
            critical_found = symptoms & critical_symptoms
            reason.append(f"Critical symptoms: {', '.join(critical_found)}")
        if temp > 39.0 and hr > 120:
            reason.append(f"High fever + tachycardia")
    
    elif temp > 38.5 or hr > 100 or bp_sys > 160:
        priority = 2
        category = "URGENT"
        reason = []
        if temp > 38.5:
            reason.append(f"Fever: {temp}°C")
        if hr > 100:
            reason.append(f"Elevated HR: {hr} bpm")
        if bp_sys > 160:
            reason.append(f"High BP: {bp_sys} mmHg")
    
    elif len(symptoms) > 0:
        priority = 3
        category = "STANDARD"
        reason = [f"Symptoms present, vitals stable"]
    
    else:
        priority = 4
        category = "NON-URGENT"
        reason = ["Minor complaint, stable vitals"]
    
    return {
        "id": patient_id,
        "priority": priority,
        "category": category,
        "reason": "; ".join(reason),
        "symptoms": symptoms,
        "vitals": vitals
    }

def generate_triage_queue(patients, critical_symptoms):

    # Sort by priority (lower number = higher priority)
