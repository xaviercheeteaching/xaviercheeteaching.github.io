## 1. Introduction

Welcome to the first and most critical step of any data analysis workflow: getting your data into your R environment. In clinical and biomedical research, data often comes from diverse sources like lab instruments, electronic health records (EHRs), or collaborators as simple text files—most commonly CSVs or TXT files. In this section, you will learn the essential skill of loading these clinical and metadata files correctly and reproducibly. We will explore the modern, recommended approach using the `readr` package for a robust workflow and also cover the legacy base R methods you will inevitably encounter in older scripts. Mastering this foundational skill is critical to ensuring data integrity from the very start of your analysis.

## 2. Key Concepts and Definitions

*   **CSV (Comma-Separated Values)**: A plain text file format that uses commas to separate distinct values in a data table. In a clinical setting, a CSV file might contain patient vitals from a monitoring device, with each row representing a time point and each column representing a different vital sign (e.g., heart rate, blood pressure).
*   **TSV (Tab-Separated Values)**: Similar to a CSV, but uses a tab character to separate values. This format is very common in bioinformatics for exporting large datasets like gene expression matrices, as it avoids conflicts with data that might contain commas (e.g., text notes).
*   **Tibble**: The modern data frame format used by the `tidyverse`, including `readr`. A key medical advantage is that tibbles do not automatically convert character strings (like a patient ID "007-PATIENT") into statistical factors, preventing a common source of errors.
*   **Factor**: An R data type used to store categorical variables (e.g., "Male"/"Female", "Control"/"Treatment"). In older versions of R, base functions would automatically convert any text column into a factor, which could corrupt data like patient identifiers or free-text notes by treating them as fixed categories.
*   **Medical Record Number (MRN)**: A unique identifier assigned to a patient by a healthcare organization. These are often used as patient IDs in datasets and must be handled as text to prevent data corruption (e.g., losing leading zeros).
*   **Reproducible Path**: A file path that works regardless of the user or computer running the script. Using a tool like the `here` package to build paths from the R Project's root directory is crucial for collaborative research, ensuring a script written on your machine runs correctly on a colleague's machine without modification.

## 3. Main Content

### 3.1 The Modern Workflow: Importing with `readr`

For a fast, reliable, and reproducible workflow, the standard recommendation is to use the `readr` and `here` packages. `readr` is part of the `tidyverse` and is optimized for performance and predictability.

```r
# Important: Run install commands ONLY in the console, not inside a script. This is a one-time setup.
# install.packages(c("readr", "here"))
library(readr)
library(here)

# Create a reproducible path and import a CSV file containing patient vital signs (e.g., Heart Rate, Blood Pressure)
patient_vitals <- read_csv(here("data", "vitals.csv"))
patient_vitals
```
```
# Output:
# # A tibble: 3 × 4
#   patient_id visit_date  heart_rate blood_pressure
#   <chr>      <date>           <dbl> <chr>
# 1 PT001234   2024-01-15          72 120/80
# 2 PT005678   2024-01-16          85 130/85
# 3 PT009101   2024-01-17          68 115/75
```

*   `readr::read_csv()` imports data as a tibble. This is highly beneficial as it prevents text columns, such as patient identifiers (e.g., Medical Record Numbers or MRNs), from being mistakenly converted into statistical factors.

> **In Practice:** Patient IDs are the primary key for merging different clinical datasets, such as linking patient vitals to lab results or medication records. If IDs are incorrectly imported as factors or numbers, these critical merge operations can fail or, worse, produce silently incorrect results. Using `readr` to ensure IDs remain as character data is a foundational step for ensuring data integrity in clinical research.

*   `here::here()` builds a file path starting from your RStudio Project folder. This simple function is incredibly powerful because it eliminates "No such file or directory" errors that occur when scripts are moved between computers or run by different users.

> **Debug Note:** The `"No such file or directory"` error is one of the most common frustrations in R. It's almost always caused by the script not knowing where to look for the file. The `here()` function solves this by creating a reliable path that starts from your R Project's root folder, making your script portable and your file paths predictable.

*   **Other Delimiters**: Use `read_tsv()` for tab-separated files (common for gene expression data) or `read_delim()` for files with other separators (e.g., semicolons).
*   **Data Type Control**: To prevent `readr` from incorrectly guessing data types, use the `col_types` argument. This is critical for ensuring patient IDs with leading zeros (e.g., `00123`) are imported as text and not as numbers.

> **Pro Tip:** Before trying to load any file, open it in a plain text editor (like Notepad on Windows or TextEdit on Mac). This allows you to quickly check the separator (is it a comma, tab, or something else?), look for issues like extra commas in notes fields, and verify the column names in the header row. A 10-second check can save you 10 minutes of debugging.

Now that you've seen the modern, recommended workflow, let's review the base R functions. You will inevitably encounter these in older scripts or code from collaborators, so understanding them is essential for maintenance and debugging.

### 3.2 The Legacy Method: Importing with Base R

You will frequently encounter base R functions in older scripts. While functional, these methods require more careful handling of data types to avoid critical errors.

```r
# Import a CSV file using base R
patient_vitals_base <- read.csv(
  here("data", "vitals.csv"),
  stringsAsFactors = FALSE
)
head(patient_vitals_base) # Using head() for base R data.frame is common
```
```
# Output:
#   patient_id visit_date heart_rate blood_pressure
# 1   PT001234 2024-01-15         72         120/80
# 2   PT005678 2024-01-16         85         130/85
# 3   PT009101 2024-01-17         68         115/75
```

> **Important:** In R versions before 4.0.0, `read.csv()` and `read.delim()` automatically converted text columns into a statistical data type called a factor. This behavior silently corrupts data like patient identifiers (e.g., MRNs like "007" become "7") and can cause catastrophic errors in data merging and analysis. Setting `stringsAsFactors = FALSE` is essential to prevent this. While modern R (4.0.0+) no longer does this by default, explicitly including this argument ensures your code is safe and reproducible for all collaborators, regardless of their R version.

*   For tab-separated files (`.txt`, `.tsv`), the base R equivalent is `read.delim()`. You must also use the `stringsAsFactors = FALSE` argument with it.

## 5. Practical Applications

*   **Cohort Assembly for Clinical Trials**: Researchers frequently receive patient eligibility and demographic data in separate CSV and TXT files from different hospital departments. Using `read_csv()` and `read_tsv()` with `here()` allows them to build a reproducible script that combines these files to assemble the final study cohort. The real-world impact is accurate patient selection and the prevention of data corruption that could compromise the entire trial.
*   **Processing High-Throughput Genomics Data**: In precision oncology, RNA-sequencing produces massive gene expression matrices, which are typically stored as large, tab-separated text files. The `readr::read_tsv()` function is highly optimized to read these multi-gigabyte files quickly and efficiently. This enables the rapid identification of differentially expressed genes between tumor and healthy tissue, guiding personalized treatment strategies.
*   **Longitudinal Analysis of Patient Vitals**: Data from Intensive Care Unit (ICU) monitoring equipment is often exported as enormous CSV files containing time-series measurements (heart rate, blood pressure, etc.). Using `read_csv()` with specified `col_types` is essential to correctly parse timestamps, patient IDs, and numeric readings. This allows data scientists to build models that track patient health trajectories, predict clinical events like septic shock, and ultimately improve patient outcomes.
