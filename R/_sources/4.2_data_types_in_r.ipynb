# Chapter 4: Data Types in R

## 1. Introduction

Welcome to the cornerstone of data handling in R: creating variables and assigning them the correct data types. This skill is critical in a medical context, where storing a patient's age as a number and their unique identifier as text is fundamental for preventing errors in analysis. In R, you assign variables with the `<-` operator.

> **Pro Tip:**
> While both `<-` and `=` can be used for assignment, the R community style guide strongly recommends using `<-`. This convention improves code readability and avoids confusion with the `==` operator used for comparison, which we will cover later.

Using the correct data type is essential, as you cannot perform math on text. An incorrect type will lead to an error.

```R
patient_id <- "PT000102"
print(patient_id) # [1] "PT000102"

# The following line causes an error because patient_id is text.
# "PT000102" + 5 
# Error in "PT000102" + 5 : non-numeric argument to binary operator
```

> **Debug Note:**
> The error "non-numeric argument to binary operator" is very common. It occurs when you try to perform a mathematical operation on a `character` (text) variable. Always check the data type with `class()` if you encounter this error.

By the end of this section, you will be able to correctly store different kinds of medical data—from lab values to patient consent status—setting a solid foundation for all future data analysis tasks.

## 2. Key Concepts and Definitions

*   **Variable**: A named storage location in memory that holds a value. In a clinical context, think of it as a label for a piece of patient data, like `patient_name` or `heart_rate`.
*   **Data Type**: The classification of a value that tells the computer how to interpret it. Correctly classifying data as a number, text, or logical value is essential for accurate analysis.
*   **Numeric**: The default data type for any number in R, including those with decimals. It is ideal for continuous measurements like a patient's temperature (`37.5`) or a drug's concentration in blood.
*   **Integer**: A data type for whole numbers only. It is used for countable data, such as the number of patients in a trial (`50L`) or the number of days since a diagnosis. Appending `L` to a number explicitly makes it an integer.
*   **Character**: The data type for text, also known as a "string." It is used to store qualitative information that is not used in mathematical calculations, such as a patient's blood type (`"O+"`) or a gene name (`"BRCA1"`).
*   **Logical**: A data type that can only hold one of two values: `TRUE` or `FALSE`. It is essential for representing binary states, such as whether a patient has provided consent (`has_consent <- TRUE`) or if a specific gene mutation is present.

## 3. Main Content

### 3.1 The Basic Data Types

**Numeric**
The `numeric` type handles numbers with decimals, making it the workhorse for continuous clinical measurements.

```R
patient_bmi <- 29.5
class(patient_bmi) # [1] "numeric"
```

> **In Practice:**
> Values like patient BMI (29.5 kg/m²), temperature (37.2 °C), or blood glucose levels (5.5 mmol/L) are all stored as `numeric` because they can have fractional parts.

**Integer**
The `integer` type is for whole numbers. You must add an `L` suffix to create one.

```R
# R defaults to numeric without the 'L'.
class(50) # [1] "numeric"

# Use 'L' to create an integer.
patient_count <- 50L
class(patient_count) # [1] "integer"
```

> **Pro Tip:**
> Always use the `L` suffix (e.g., `50L`) to create an `integer`. While R can often use `numeric` values like `50.0` where an integer is needed, some functions for counting or indexing require true integers. Explicitly creating an integer makes your code's intent clearer and prevents potential errors.

**Character**
The `character` type stores text (strings) and is essential for qualitative and categorical information.

```R
blood_type <- "O+"
class(blood_type) # [1] "character"
```

> **In Practice:**
> `character` data is used for patient identifiers (`"PT000102"`), blood types (`"O+"`), gene names (`"BRCA1"`), or notes from a patient file.

**Logical**
The `logical` type stores `TRUE` or `FALSE` values, acting as binary flags in medical datasets.

```R
has_consent <- TRUE
class(has_consent) # [1] "logical"
```

> **In Practice:**
> `logical` values are perfect for tracking statuses like patient consent (`has_consent <- TRUE`), whether a patient is a smoker (`is_smoker <- FALSE`), or if a specific gene mutation is present.

### 3.2 Calculations with Variables

You can perform mathematical operations directly on numeric variables.

> **Important:**
> This calculation is a simplified example for teaching purposes **ONLY**. Real-world dosage calculations are far more complex and must be performed by qualified medical professionals using validated clinical tools. **NEVER** use this sample code for actual clinical decisions.

> **Medical Background:**
> The dosage for acetaminophen in children is typically calculated based on weight, commonly 10-15 mg per kg. This example uses a weight of 15 kg and a dosage of 15 mg/kg. (Source: Kliegman et al., *Nelson Textbook of Pediatrics*, 2020).

```R
patient_weight_kg <- 15   # A 15 kg child's weight
DOSAGE_MG_PER_KG <- 15     # Standard dose constant

total_dosage_mg <- patient_weight_kg * DOSAGE_MG_PER_KG

# Use print() to explicitly display output, necessary when running a script.
print(total_dosage_mg)     # [1] 225
```

### 3.3 Automatic Type Coercion

R converts data types during operations to prevent data loss.

```R
patient_age_int <- 67L      # An integer, e.g., patient's age in years
adjustment_factor <- 0.5    # A numeric, e.g., a mid-year adjustment
adjusted_age <- patient_age_int + adjustment_factor

class(adjusted_age) # [1] "numeric"
print(adjusted_age) # [1] 67.5
```

> **Pro Tip:**
> R performs automatic type coercion to prevent data loss. When you combine an `integer` and a `numeric`, R chooses `numeric` for the result because converting 29.5 to an integer (29) would lose the decimal information. This hierarchy generally follows: `logical` → `integer` → `numeric` → `character`.

## 5. Practical Applications

*   **Managing Patient Demographics:** In electronic health records (EHRs), patient data is stored using various data types. A patient's name and address are stored as `character` data, their age as an `integer`, and their weight or height as `numeric` values. Correctly typing this data ensures that you can calculate a patient's BMI (`numeric`) while preventing mathematical errors on their name (`character`).
*   **Analyzing Clinical Lab Results:** Lab panels generate a mix of data. A blood glucose level of `98.6` is `numeric`, while a test result flag like `is_abnormal` would be `logical` (`TRUE` or `FALSE`). Storing these correctly allows researchers to quickly filter for all patients with abnormal results and then perform statistical calculations on their corresponding numeric lab values.
*   **Tracking Clinical Trial Consent:** In clinical trials, managing patient consent is a critical ethical and legal requirement. A `logical` variable, such as `has_signed_consent <- TRUE`, provides an unambiguous, machine-readable record. This allows for automated checks to ensure that only data from consented patients is included in analyses, safeguarding patient rights and data integrity.
