## 2. Key Concepts and Definitions

*   **Faceting**: The process of splitting a single plot into a matrix of subplots based on the levels of one or more categorical variables. This is a core technique for subgroup analysis. In a clinical trial, faceting allows you to create separate plots of drug response for each `study_arm` and demographic group, all within one figure.
*   **Panel (or Facet)**: A single subplot within a larger faceted plot. Each panel displays the same type of visualization but for a different subset of the data. For example, one panel might show a scatter plot for male patients, while another panel shows the same plot for female patients.
*   **`facet_wrap()`**: A `ggplot2` function used to create a series of subplots (panels) based on a single categorical variable. The panels are typically "wrapped" into a grid. This is ideal for comparing serum creatinine distributions across multiple study arms arranged in a flexible layout.
*   **`facet_grid()`**: A `ggplot2` function that creates a rigid 2D grid of panels, mapping one variable to the rows and another to the columns. This is useful for visualizing interactions between two variables, such as observing how a biomarker changes conditional on both `gender` (rows) and `study_arm` (columns).
*   **Plot Composition**: The technique of arranging multiple, independent `ggplot` objects into a single, complex figure. This goes beyond faceting, as the plots can be of entirely different types (e.g., a boxplot, a scatter plot, and a map). The `patchwork` package simplifies this process, enabling the creation of comprehensive diagnostic or research dashboards.

## 3. Main Content

This section covers the primary functions for creating multi-panel plots. We will use a sample dataset simulating serum creatinine levels for patients in a renal drug trial. This data will serve as the foundation for all examples.

```R
# Load required packages
library(ggplot2)
library(patchwork)

# Sample data: serum creatinine levels for patients in a renal drug trial
set.seed(42)
patient_data <- data.frame(
  study_arm = rep(c("A", "B", "C"), each = 20),
  gender = rep(c("M", "F"), times = 30),
  age = sample(30:75, 60, replace = TRUE),
  creatinine_level_mg_dl = c(rnorm(20, 1.3, 0.2), rnorm(20, 2.6, 0.3), rnorm(20, 0.7, 0.2))
)
```

### 3.1 `facet_wrap()`: Subplots for One Variable

Use `facet_wrap()` to create a series of plots for each level of a single categorical variable. This is your go-to function for comparing subgroups along one dimension.

> **In Practice:**
> Imagine you're analyzing patient satisfaction scores across 12 different hospital wards. Instead of creating 12 separate plots, you can use `facet_wrap(~ ward)` to instantly generate a grid of plots. This makes it easy to compare performance and spot which wards are outliers without manually filtering the data for each one.

```R
# Split plots by study_arm into 3 columns
ggplot(patient_data, aes(x = age, y = creatinine_level_mg_dl)) +
  geom_point() +
  facet_wrap(~ study_arm, ncol = 3) +
  labs(title = "Creatinine Level by Age, Faceted by Study Arm")
# Expected Output: A figure containing three side-by-side scatter plots (in a single row).
# Each plot corresponds to a study arm ('A', 'B', 'C') and shows
# creatinine_level_mg_dl vs. age for that group.
```

> **Pro Tip:**
> Use the `scales` argument when data ranges vary significantly across facets. For example, `scales = "free_y"` allows the y-axis of each subplot to adapt to its own data range. This prevents plots with smaller value ranges from being compressed and unreadable.

### 3.2 `facet_grid()`: Subplots for Two Variables

Use `facet_grid()` to create a rigid 2D grid of plots based on two variables. This is excellent for exploring interactions between two categorical factors.

> **Medical Background:**
> In clinical trials, results are often stratified by both a treatment arm and a demographic factor (e.g., gender). `facet_grid(gender ~ study_arm)` creates a structured matrix that is perfect for this analysis, mirroring the layout of summary tables commonly found in medical research papers.

```R
# Map 'gender' to rows and 'study_arm' to columns
ggplot(patient_data, aes(x = age, y = creatinine_level_mg_dl)) +
  geom_point() +
  facet_grid(gender ~ study_arm) +
  labs(title = "Creatinine Level by Age, Faceted by Gender and Study Arm")
# Expected Output: A figure containing a 2x3 grid of scatter plots. The rows
# represent gender ('F', 'M') and the columns represent study arms ('A', 'B', 'C').
```

### 3.3 `patchwork`: Combine Different Plots

Use the `patchwork` package to combine entirely different `ggplot` objects into one composite view. This is essential for building dashboards that show a high-level summary (like a boxplot) next to a detailed view (like a scatter plot).

```R
# Plot 1: Scatter plot of overall trend
p1 <- ggplot(patient_data, aes(x = age, y = creatinine_level_mg_dl)) +
  geom_point() +
  labs(title = "Overall Trend")

# Plot 2: Box plot of distribution by group
p2 <- ggplot(patient_data, aes(x = study_arm, y = creatinine_level_mg_dl)) +
  geom_boxplot() +
  labs(title = "Distribution by Study Arm")

# Place plots side-by-side and add a main title
(p1 + p2) + plot_annotation(title = "Overall and Group-Specific Creatinine Trends")
# Expected Output: A figure with two plots side-by-side under a single main title.
# On the left, a scatter plot of age vs. creatinine level. On the right, a box plot
# showing the distribution of creatinine level for each study arm.
```

## 5. Practical Applications

*   **Clinical Trial Subgroup Analysis**: During a clinical trial, researchers must assess if a new drug is effective across different patient subgroups. They can use `facet_wrap(~ demographic_variable)` to create plots of efficacy (e.g., tumor size reduction) for each age group, gender, or genetic marker, quickly identifying if the treatment benefits some populations more than others.
*   **Longitudinal Patient Monitoring**: A clinician tracking a patient with a chronic disease might want to see how multiple biomarkers have changed over time. They could use `facet_grid(biomarker_name ~ ., scales = "free_y")` to create a vertical stack of plots, with each panel showing the trend for a different biomarker (e.g., HbA1c, creatinine, blood pressure) for that single patient, providing a comprehensive historical view.
*   **Integrated Diagnostic Dashboard**: For a complex case, a physician needs to synthesize information from different sources. Using `patchwork`, an analyst can create a dashboard that combines a patient's ECG trace (a line plot), their latest bloodwork results (a bar or dot plot), and their vital signs history (a scatter plot). This composite view, created with `(p_ecg / p_vitals) + p_bloodwork`, gives a holistic and immediate understanding of the patient's condition.
