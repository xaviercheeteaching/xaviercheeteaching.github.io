# Chapter 6: Essential File and Directory Operations

## 1. Introduction

Welcome to file searching on the command line, a fundamental skill for managing the vast datasets common in precision health. This section introduces two of the most powerful tools in your arsenal: `grep` and `find`. The key distinction is simple: `grep` finds *content* inside files (the "what"), while `find` locates the *files themselves* based on attributes like name or date (the "where"). Our learning journey will cover how to use each tool individually, from simple searches to complex pattern matching, and then how to combine them to perform sophisticated data discovery. Mastering these commands will enable you to efficiently query, manage, and validate the thousands of data files encountered in any large-scale bioinformatic analysis.

> **In Practice:** Imagine you're a clinical researcher with thousands of patient data files. You might use `grep` to quickly find every file that mentions a specific gene mutation ("the what"). Then, you could use `find` to locate all of those files that were generated in the last week to prepare them for a new analysis ("the where"). Mastering both tools is essential for efficient data exploration.

---

## 2. Key Concepts and Definitions

*   **`grep` (Global Regular Expression Print)**: A command-line utility that searches for a specific text pattern within files. In a medical context, you might use `grep` to scan a patient's electronic health record (EHR) file for every mention of the drug "Metformin."
*   **`find`**: A command-line utility that locates files and directories based on their attributes, such as name, size, type, or modification date. As an analogy, while `grep` reads the books in a library, `find` locates all books published in the last year or all books with "Genetics" in the title.
*   **Pattern**: The specific sequence of characters or text that `grep` searches for. A pattern can be a simple word like "cancer" or a complex regular expression that matches a patient ID format.
*   **Regular Expression (Regex)**: A special sequence of characters that defines a search pattern, allowing for more flexible and powerful searches than simple text. For example, a regex could find all gene IDs that start with "BRCA" followed by any single digit (e.g., `BRCA1`, `BRCA2`).
*   **File Attribute**: A piece of metadata about a file that `find` can use as a search criterion. Common attributes include the filename, file size, modification time, and file type (e.g., file or directory).
*   **Recursive Search**: A search that traverses a directory and all of its subdirectories. This is crucial in bioinformatics, where data is often organized in nested folders, such as `/project/patient_id/sequencing_run/`.

> *   `-n`: "less than n" (e.g., `-mtime -2` means less than 2 days old).

*   **Safely delete old backup files with `-delete`.**

## 5. Practical Applications

*   **Pharmacogenomics Screening**: A researcher can use `grep -w "CYP2D6"` on thousands of variant call format (VCF) files to quickly identify patients with variants in a key drug-metabolizing gene. This helps shortlist candidates for a pharmacogenomics study without manually opening each file.
*   **Data Archiving and Cleanup**: A systems administrator at a hospital can use `find /data/ehr_backups/ -mtime +365 -delete` to automatically locate and remove electronic health record backups that are more than a year old, ensuring compliance with data retention policies and freeing up storage space.
*   **Clinical Trial Data Validation**: During a clinical trial, a data manager can run `grep -rli "error|warning" /data/trial_XYZ/` to recursively search all data submission folders for any files containing the words "error" or "warning". This provides a quick quality control check to identify problematic data files that need investigation.
*   **Identifying Failed Sequencing Runs**: A bioinformatician can use `find /sequencing_data/ -type f -name "*.fastq.gz" -empty` to locate any empty FASTQ files. An empty file often indicates a failed data transfer from a sequencing machine, allowing for rapid identification and re-initiation of the transfer.

## 6. Summary and Key Takeaways

In this section, we've explored the essential command-line tools for searching: `grep` for finding content *within* files and `find` for locating the *files themselves*. We saw how to use various options to refine our searches and how to combine these tools to perform complex data discovery tasks critical for managing large-scale precision health data.

*   **`grep` is for content**: Use it to find lines matching a text pattern (the "what"). Key options include `-i` (ignore case), `-c` (count), `-v` (invert), `-l` (list files), and `-r` (recursive).
*   **`find` is for files**: Use it to locate files based on attributes like name, size, and modification time (the "where"). Key tests include `-name`, `-type`, `-mtime`, and `-size`.
*   **Combine for Power**: You can use `find ... -exec grep ... {} +` to first locate a specific set of files and then search for content within them.
*   **Safety First**: When using actions like `-delete` or `-exec` with `find`, always run the command without the action first to ensure you are targeting the correct files.
