## 1. Introduction

Welcome to the core of scripting automation: variables. In precision health, we are constantly handling data—patient IDs, gene names, sequencing read counts, and lab values. Manually entering this information is slow and prone to error. Variables act as named containers, allowing us to store, retrieve, and manipulate this data efficiently. ## 2. Key Concepts and Definitions

Understanding these core concepts is crucial for using variables effectively in your scripts. Each term represents a building block for creating dynamic and reliable automations.

*   **Variable**: Think of a variable as a labeled container or a sticky note on a lab sample. The label is the variable's name (e.g., `patient_id`), and the content inside is its value (e.g., `"PT9876543A"`). This allows you to refer to data by a memorable name instead of the raw value itself.
*   **String**: A sequence of characters treated as text. Patient identifiers, filenames, and gene symbols are all stored as strings. In Bash, strings are often enclosed in quotes (e.g., `"BRCA1"`).
*   **Integer**: A whole number (without a decimal point). Integers are essential for counting operations, such as tallying the number of genetic variants found in a sample or counting the total samples in a sequencing run.
*   **Hemoglobin (Hb)**: A protein in red blood cells that carries oxygen from the lungs to the body's tissues. Its concentration is a common blood test parameter used to screen for conditions like anemia.
*   **Body Mass Index (BMI)**: A value derived from the mass and height of a person, used as a screening tool to identify potential weight-related health issues.
*   **Read-Only Variable**: A variable whose value is "locked" and cannot be changed after it is set. This is used to protect critical, constant information, like the file path to a human reference genome, from being accidentally overwritten during script execution.
*   **Command Substitution**: A powerful feature that allows you to capture the output of a command and store it in a variable. For example, you could run a command to count the lines in a file (`wc -l`) and store that count directly into a variable named `line_count`.

### 3.4. Create Read-Only Variables

While variables are designed to be flexible, some data is too important to change. Let's see how to protect it. Make a variable read-only with `readonly` to protect critical data, like a reference file path, from being overwritten.

````python
%%bash
# Declare a variable as read-only.
readonly GENOME_FILE="/data/hg38/chr1.fa"
# This will now cause an error and stop the script.
GENOME_FILE="/new/path/file.fa"
````

### 3.6. Command Substitution

So far, we've assigned static values or simple math results to variables. A more powerful technique is to assign the output of an entire command to a variable. Use command substitution `$(...)` to capture a command's output.

````python
%%bash
# Calculate BMI from weight (kg) and height (m).
patient_weight_kg=70.5
patient_height_m=1.75
# (Formula Source: CDC 2023)
# Use the bc (basic calculator) utility for floating-point math; bash cannot do this natively.
bmi=$(echo "scale=2; $patient_weight_kg / $patient_height_m^2" | bc)
echo "Patient BMI: $bmi"
````

## 5. Practical Applications

Variables are indispensable in computational biology and precision health. They serve as the glue that connects different tools and automates complex workflows.

*   **Dynamic File Naming in NGS Pipelines**: In a genomics workflow, a script can take a patient ID as a variable. All subsequent output files—from alignment (`.bam`) to variant calls (`.vcf`)—can be automatically named using this variable (e.g., `${patient_id}.bam`). This makes the pipeline scalable and eliminates manual naming errors when processing hundreds of samples.
*   **Centralized Configuration Management**: Scripts for complex analyses often have many parameters, such as file paths to reference genomes, annotation databases, or software executables. Storing these as `readonly` variables at the top of a script (e.g., `readonly REF_GENOME="/data/genomes/hg38.fa"`) makes the script easier to read, maintain, and configure for different computing environments.
*   **Automated Quality Control Reporting**: A script can run a tool like FastQC, then use command substitution to `grep` for key metrics like "Per base sequence quality" from the output log. These metrics can be stored in variables and compared against thresholds or printed into a concise, custom summary report for a lab manager.
