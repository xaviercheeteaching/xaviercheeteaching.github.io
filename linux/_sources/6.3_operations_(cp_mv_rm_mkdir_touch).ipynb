## 2. Key Concepts and Definitions

*   **Directory**: A folder used to organize files. In a clinical setting, a directory for a patient (e.g., `PT007890`) is like that patient's individual medical chart, holding all their related documents.
*   **File**: A container for storing information, such as text, data, or a program. A file like `20251111_CBC_results.txt` holds specific data for a patient.
*   **Path**: The unique location of a file or directory in a file system. A path like `/data/wgs/PT007890/raw_reads/` is the precise "address" of a patient's genetic information.
*   **Flag (or Option)**: A modifier, like `-p` or `-r`, that changes a command's behavior. This is like giving a specific instruction, such as telling a filing clerk to "copy the *entire* patient file, including all sub-folders," not just the cover sheet.
*   **WGS (Whole Genome Sequencing)**: A laboratory process that determines the complete DNA sequence of an organism's genome. The resulting data files are often very large and require careful management.
*   **CBC (Complete Blood Count)**: A common blood test used to evaluate overall health. The results are often stored in text or structured data files.

> **Debug Note:** Remember that `mv` serves two functions. If the target is a directory, it moves the file(s). If the target is a new name in the *same* directory, it renames the file. ### 3.5 `rm` and `rmdir`: Remove Files and Directories

These commands permanently delete files and directories. They must be used with extreme caution.

> **Important:** The `rm` command permanently deletes files without moving them to a trash bin. This action is generally unrecoverable. The `rm -r` command is especially powerful as it deletes entire directories and all their contents. Always double-check your command before executing it, and use the `-i` flag for an extra layer of safety.

````python
%%bash
# Safely remove a single file with a confirmation prompt
rm -i temp_notes.txt
# Safely remove an empty directory only
rmdir empty_folder/
# Recursively remove a directory and its contents (USE WITH EXTREME CAUTION)
rm -r old_project_data/
````

## 5. Practical Applications

*   **Setting Up a New Patient Project**: A clinical bioinformatician receives a new tumor sample for sequencing. They use `mkdir -p PT0123/{raw_data,analysis,reports}` to instantly create a standardized, organized directory structure. This ensures all subsequent data and results are stored in a predictable and reproducible manner.
*   **Applying a Standard Analysis Pipeline**: A research lab has a validated R script (`somatic_variant_calling.R`). When a new dataset arrives, they use `cp templates/somatic_variant_calling.R PT0456/analysis/` to copy the script into the new project. This prevents accidental modification of the original template and ensures consistency across analyses.
*   **Data Curation and Renaming**: Data from a sequencing core arrives with generic names like `LANE1_S1_R1.fastq.gz`. A technician uses a script with the `mv` command to rename it to `PT0789_TUMOR_20251111_R1.fastq.gz`, embedding critical metadata directly into the filename for better tracking.
*   **Managing Storage by Cleaning Up Intermediate Files**: A whole-genome alignment process can generate terabytes of temporary files. After the final, validated results are created, the bioinformatician uses `rm` to delete the large intermediate files, freeing up critical space on the high-performance computing cluster.
