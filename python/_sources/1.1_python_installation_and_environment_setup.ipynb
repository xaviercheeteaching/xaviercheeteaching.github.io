## 2. Key Concepts and Definitions

Understanding these core terms is essential for managing your analytics workspace effectively.

*   **Python**: A versatile, high-level programming language renowned for its readability and extensive libraries. In precision health, it is the workhorse for everything from processing genomic data and analyzing clinical trial results to building predictive diagnostic models.
*   **Virtual Environment (`venv`)**: An isolated, self-contained directory tree that contains a specific Python installation and its own set of libraries. In a medical context, this is like having a dedicated, sealed-off lab for each experiment. It guarantees that your analysis of "Patient Cohort A" using `pandas` version 1.5 is completely unaffected by a different project that requires `pandas` version 2.0, preventing cross-contamination of your digital tools.
*   **PATH Variable**: An environment variable on your operating system that specifies a set of directories where executable programs are located. When you type a command like `python` in your terminal, the system searches these directories to find the program. Adding Python to your PATH during installation makes it universally accessible, which is critical for running analysis scripts from any location.
*   **pip (Pip Installs Packages)**: The standard package manager for Python. It allows you to install, update, and manage the third-party libraries and dependencies your project needs. For a clinical data scientist, `pip` is the tool used to procure specialized instruments like `NumPy` for numerical analysis, `SciPy` for statistical tests, and `Scikit-learn` for machine learning.
*   **Execution Policy (PowerShell)**: A security feature in Windows PowerShell that determines whether scripts can be run. By default, it's often restricted. Setting it to `RemoteSigned` for the current user is a necessary one-time configuration that allows you to run trusted local scripts, such as the virtual environment activation script, without compromising system security.

### 3.1 Checking Python environment in Jupyter

First, to confirm that Jupyter-Lite has Python installed, run the command below:

```python
import sys
print(sys.version)
```

### 3.2 Checking if a Package exist

A **Python package** is a collection of related code that someone has bundled together to perform specific tasks. Think of it as a toolkit you can add to Python to extend its capabilities.

You can try and import Ipywidgets, an important package that makes interactive HTML widgets for Jupyter notebooks that make notebooks dynamic and user-friendly.

```python
import ipywidgets
print(f"Ipywidgets is available!")
print(f"Version: {ipywidgets.__version__}")
```

### 3.3 Installing and importing a package

Now, let's try and import another Python package called Plotly, an interactive browser-based graphing library for Python.  

```python
import plotly
```

Oh no! You have run into an error. 
The error tells you that the plotly module is either not **installed** or **imported** in Jupyter-lite. 

> There is a different between **installed** and **imported**
>
> The simplest way to think about it is using an analogy: a toolbox and a workbench.
> 
> * **Installing** is like buying a new tool (e.g., a specific wrench) and placing it in your toolbox. The tool is now in your possession and available for you to use whenever you need it.
> 
> * **Importing** is like taking that specific wrench out of the toolbox and placing it on your workbench to use for your current project.
> 
> You only need to install the tool once. But, every time you start a new project (or run a new script), you need to import the tools you need for that specific job.

Let's first install Plotly

```python
!pip install plotly
```

> NOTE: The `!` tells Jupyter notebooks/IPython to run the command in your system's terminal/shell instead of treating it as Python code—in a regular terminal, you'd just use pip install plotly without the `!`.

Once installed, now you can import Plotly and check the version that you are running. 

```python
import plotly
print(f"Plotly is available!")
print(f"Version: {plotly.__version__}")
```

Hooray! You've installed Plotly onto your Jupyter-lite environment!
