## 2. Key Concepts and Definitions

*   **DataFrame**: The primary data structure in Pandas, analogous to a digital spreadsheet or a patient's medical chart. Each row represents an observation (e.g., a single clinic visit), and each column represents a variable (e.g., heart rate, temperature).
*   **CSV (Comma-Separated Values)**: A plain text file format used to store tabular data. Each line is a data record, and each record consists of one or more fields, separated by commas. In a medical context, a daily export of patient vitals from a monitoring device is often in CSV format.

### 3.2 Editing CSV files

Now that we know how to read data, let's look at how to add new rows to existing csv file.

```python

filename = 'vitals_data.csv'

# The new row we want to add (ensure the order matches the columns!)
new_row = ['PT000008', 55, 78, 37.2]

# Open the file in append mode ('a')
# newline='' is required to prevent blank lines between rows on some systems
with open(filename, mode='a', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(new_row)

print(f"Added new patient to {filename}.")

# Verify the result by reading it back
updated_df = pd.read_csv(filename)
print(updated_df.tail()) # .tail() shows the last few rows
```

### 3.3 Deleting Rows in CSV files

To delete a row, you typically load the data, remove the specific row(s), and then save the file again. We rarely delete rows directly from the raw file on the disk because it is error-prone.

Here are the two most common ways to do this using pandas.

**Method 1: Drop by Index (Row Number)**
This is useful if you know the exact position of the row you want to delete (e.g., "delete the 3rd row").

```python
import pandas as pd

# 1. Load the original file
df = pd.read_csv('vitals_data.csv')

# 2. Delete the row with index 2 (which is the 3rd row, PT000003)
# Note: simple drop() returns a NEW dataframe, it doesn't change the original one unless you say so
df_cleaned = df.drop(index=2)

# 3. Save the result back to CSV
# We overwrite the original file, or you could save to a new name
df_cleaned.to_csv('vitals_data_v2.csv', index=False)
updated_df = pd.read_csv('vitals_data_v2.csv')
print(updated_df)
```

**Method 2: Drop by Condition (Filtering)**

This is the more robust "data science" way. You keep only the rows that match a specific rule. For example, "keep all patients where Age is not 54".

```python
import pandas as pd

# 1. Load the file
df = pd.read_csv('vitals_data.csv')

# 2. Filter: Keep rows where 'patient_id' is NOT 'PT000002'
df_filtered = df[df['patient_id'] != 'PT000002']

# 3. Save the result
df_filtered.to_csv('vitals_data_3.csv', index=False)
updated_df = pd.read_csv('vitals_data_3.csv')
print(updated_df)
```

## 5. Practical Applications

*   **Aggregating Clinical Trial Data:** Researchers often receive patient data from multiple trial sites as separate CSV files. Using a loop and `pd.read_csv()`, they can ingest all files into a single master DataFrame, standardizing column names with the `names` parameter to ensure consistency for a meta-analysis.
*   **Processing Genomic Sequencer Output:** Genomic data files are often massive, tab-delimited, and prefixed with many lines of metadata. A bioinformatician would use `pd.read_csv()` with `sep='\t'`, `skiprows`, and `usecols` to efficiently parse these files, ignoring the metadata and loading only the essential gene and expression columns to conserve memory.
*   **Integrating EHR Data Exports:** When a hospital exports a large dataset from its Electronic Health Record (EHR) system, patient IDs (e.g., '00123') might be incorrectly converted to numbers (123). Using `dtype={'patient_id': str}` during import prevents this data corruption and ensures patient records can be accurately joined later.
