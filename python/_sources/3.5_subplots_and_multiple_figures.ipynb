## 2. Key Concepts and Definitions

*   **Figure**: The top-level container for all plot elements, including titles, legends, and one or more `Axes`. In a medical analogy, the `Figure` is the patient's entire medical chart file, holding all the information for a specific visit or study.
*   **Axes**: An individual plot or chart within a `Figure`. It contains the data points, x- and y-axes, labels, and title for a single visualization. In our analogy, an `Axes` object is a single page or report within the chart, such as the vital signs log or a lab results sheet.
*   **Subplot**: A common term for one of several `Axes` objects arranged in a grid within a single `Figure`. Creating subplots is the primary method for comparing related plots side-by-side.
*   **`plt.subplots()`**: The core Matplotlib function used to create a `Figure` and a grid of `Axes` (subplots) in a single command. It is the modern and recommended approach for managing multi-plot figures.
*   **`sharex` / `sharey`**: Boolean arguments for `plt.subplots()` that link the axes of different subplots. Setting `sharex=True` ensures all subplots in a column use the same x-axis ticks and labels, which is perfect for comparing different metrics over the same time period. `sharey=True` does the same for the y-axis across a row, ideal for comparing data with the same units and scale.

### 3.1 Creating Subplot Grids

Use `plt.subplots()` to create a consolidated grid of plots for comparing related data, like a patient's vital signs. The function returns a `Figure` object and a NumPy array of `Axes` objects.

```python
import matplotlib.pyplot as plt
import numpy as np

time = np.array([0, 4, 8, 12])
heart_rate = np.array([75, 78, 72, 80])
systolic_bp = np.array([120, 122, 118, 125])

# Create a 2-row, 1-column grid with a shared x-axis
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, layout='constrained')
fig.suptitle('Patient Vitals')

ax1.plot(time, heart_rate)
ax1.set_title('Heart Rate')
ax1.set_ylabel('Heart Rate (bpm)')

ax2.plot(time, systolic_bp)
ax2.set_title('Systolic BP')
ax2.set_ylabel('mmHg')
ax2.set_xlabel('Time (hours)')

plt.show()

# Expected Output: A figure window titled 'Patient Vitals' containing two plots
# stacked vertically. The top plot shows 'Heart Rate' and the bottom plot
# shows 'Systolic BP'. Both plots share the same x-axis labeled 'Time (hours)'.
```

*   `plt.subplots(2, 1, sharex=True, ...)`: Creates a 2-row, 1-column grid and links the x-axes to remove redundant labels and improve comparison.
*   `fig, (ax1, ax2) = ...`: Unpacks the figure and axes into separate variables. **Pro Tip:** This tuple unpacking is the standard, most readable way to access individual subplots. It is much clearer than accessing them by index (e.g., `axes[0]`, `axes[1]`).
*   `layout='constrained'`: Automatically adjusts spacing to prevent titles and labels from overlapping.

> **Key Terms:**
> *   **Figure:** The top-level container for everything you see on the plot, including all subplots, titles, and labels. Think of it as the entire canvas or window. You use `fig` methods for figure-wide settings (e.g., `fig.suptitle()`, `fig.supxlabel()`).
> *   **Axes:** An individual subplot within the figure. This is the object you actually plot data *on*. You use `ax` methods for subplot-specific modifications (e.g., `ax.plot()`, `ax.set_title()`).

The following example compares blood glucose levels for a patient with type 2 diabetes before and after receiving medication. The `sharey` argument is used to link y-axes for a direct comparison of values with the same unit.

```python
import matplotlib.pyplot as plt
import numpy as np

hours = np.array([0, 4, 8, 12])
glucose_pre = np.array([180, 175, 185, 190])
glucose_post = np.array([140, 130, 135, 128])

fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, layout='constrained')
fig.suptitle('Glucose Levels')

ax1.plot(hours, glucose_pre, color='gray')
ax1.set_title('Pre-Medication')
ax1.set_ylabel('Glucose (mg/dL)')

ax2.plot(hours, glucose_post, color='green')
ax2.set_title('Post-Medication')

fig.supxlabel('Hours Post Meal')

plt.show()

# Expected Output: A figure window titled 'Glucose Levels' with two plots
# side-by-side. The left plot shows 'Pre-Medication' levels and the right
# shows 'Post-Medication' levels. Both plots share the same y-axis scale.
```

### 3.2 Creating Multiple, Separate Figures

While subplot grids are perfect for comparing related data, sometimes you need to display completely unrelated charts, such as data for two different patients. For this, you should create multiple, separate figures by calling `plt.subplots()` multiple times.

```python
import matplotlib.pyplot as plt
import numpy as np

time = np.array([0, 4, 8, 12])
patient_a_hr = np.array([75, 78, 72, 80])
patient_b_hr = np.array([88, 90, 85, 92])

# Figure 1: Patient A
fig1, ax1 = plt.subplots(layout='constrained') # Creates a default 1x1 grid
ax1.plot(time, patient_a_hr)
ax1.set_title('Patient A - Heart Rate')
ax1.set_ylabel('Heart Rate (bpm)')

# Figure 2: Patient B
fig2, ax2 = plt.subplots(layout='constrained') # Creates a second, separate figure
ax2.plot(time, patient_b_hr, color='green')
ax2.set_title('Patient B - Heart Rate')
ax2.set_ylabel('Heart Rate (bpm)')

plt.show()

# Expected Output: Two separate figure windows displayed simultaneously.
# One window shows the heart rate for Patient A, and the other shows the
# heart rate for Patient B.
```

```python
import matplotlib.pyplot as plt
```

```python
months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the exercise
time_h = np.array([0, 6, 12, 18, 24])
temp_patient_X = np.array([37.1, 37.5, 38.2, 37.4, 37.0]) # Celsius

fig1.savefig('patient_X_temp.png')

# Figure 2 for Patient Y
fig2, ax2 = plt.subplots(layout='constrained')

ax2.grid(True)
fig2.savefig('patient_Y_temp.png')

# plt.show() is not needed when only saving files.
# Good practice to close figures after saving to free up memory.
plt.close(fig1)
plt.close(fig2)
```
**Explanation:** Each call to `plt.subplots()` creates a new, active figure. We assign the figure and axes objects to different variables (`fig1`/`ax1`, `fig2`/`ax2`) to keep them distinct. The `fig.savefig()` method is called on each figure object to save it to disk.

## 5. Practical Applications

*   **Longitudinal Patient Monitoring:** In an Intensive Care Unit (ICU), a patient's data is monitored continuously. Subplots are used to create a dashboard showing heart rate, blood pressure, oxygen saturation, and respiratory rate on a shared time axis. This allows clinicians to quickly identify correlations and detect adverse events, such as a drop in oxygen saturation coinciding with a rise in heart rate.
*   **Comparative Genomics and Expression Analysis:** When studying a disease like cancer, researchers might compare gene expression levels between a tumor sample and healthy tissue. A figure could use two subplots: one showing the expression levels of key oncogenes and another showing tumor suppressor genes. Using `sharey` can help visualize the magnitude of expression changes across both gene sets.
*   **Clinical Trial Subgroup Analysis:** During a clinical trial, it's important to see if a drug's effect varies by demographic. A 2x2 subplot grid could be used to show the treatment effect (e.g., change in a biomarker) for four different subgroups: males under 50, males over 50, females under 50, and females over 50. This provides a quick, comprehensive overview of the drug's performance across key populations.

## 6. Summary and Key Takeaways

In this section, we've explored how to move beyond single plots to create complex, multi-part visualizations for medical data. You learned the fundamental distinction between a `Figure` (the canvas) and `Axes` (the individual plots) and how to use `plt.subplots()` to manage them effectively.

*   **Consolidated Grids:** `plt.subplots()` is the primary tool for creating grids of plots, ideal for comparing related datasets.
*   **Axis Sharing is Key:** Use the `sharex` and `sharey` arguments to link axes for cleaner, more accurate comparisons, especially when data shares a common time frame or unit of measurement.
*   **Object-Oriented Control:** Unpacking the `Figure` and `Axes` objects (e.g., `fig, (ax1, ax2) = ...`) gives you explicit control to modify each component of your visualization independently.
*   **Separate Figures for Separate Ideas:** When data is unrelated (e.g., different patients), create multiple independent figures by calling `plt.subplots()` for each one.
