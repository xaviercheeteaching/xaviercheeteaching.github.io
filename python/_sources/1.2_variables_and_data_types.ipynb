## 2. Key Concepts and Definitions

*   **Variable**: A named storage location in a computer's memory used to hold a value. In a medical context, a variable might hold a patient's name (`patient_name`), age (`patient_age`), or a specific lab result (`serum_creatinine`).
*   **String (`str`)**: A sequence of characters used to store text data. Strings are enclosed in single (`'`) or double (`"`) quotes. They are essential for identifiers like Medical Record Numbers (MRNs) or patient names, as they preserve all characters.
*   **Integer (`int`)**: A whole number without a fractional part. Integers are used for discrete, countable data, such as a patient's age in years (`52`), the number of hospital admissions, or heart rate in beats per minute.
*   **Float (`float`)**: A number that contains a decimal point. Floats are critical for representing continuous measurements where precision is vital, such as body temperature (`37.0` °C), drug dosages (`1.25` mg), or lab values like hemoglobin A1c (`6.5` %).
*   **Boolean (`bool`)**: A data type that can only have one of two values: `True` or `False`. Booleans are perfect for representing binary states in clinical data, such as an allergy status (`is_allergic_to_penicillin = True`) or the presence of a comorbidity (`has_diabetes = False`).
> **Debug Note:** A common mistake for new programmers is typing `true` or `false` in lowercase. Python is case-sensitive, so these must be capitalized (`True`, `False`). Using lowercase will result in a `NameError` because Python will think you are trying to use a variable that hasn't been defined.
*   **Type Casting**: The process of converting a variable from one data type to another (e.g., converting a string `"14.5"` to a float `14.5`). This is a common task when processing data imported from text files or databases.

### 3.1 Naming Conventions and Best Practices

To ensure your code is readable and maintainable, especially in collaborative medical research projects, it's crucial to follow standard naming conventions.

> **Pro Tip:** Use `snake_case` for variable names (e.g., `patient_age_years` instead of `patientageyears` or `PatientAgeYears`). This is the standard convention in Python (see PEP 8 style guide) and makes your code significantly more readable for you and your colleagues.

Additionally, adding comments to note units for clinical measurements is a vital best practice that prevents misinterpretation.

```python
# Good practice: specify units in a comment
```

### 3.2 Verifying and Converting Data Types

**Changing a Type (Casting)**
You can convert data between compatible types using the `str()`, `int()`, and `float()` functions. This is often required when a lab value is read from a file as text but is needed as a number for analysis.

```python
# Lab result read from a file as a string
hemoglobin_g_dl = "14.5" # Unit: g/dL
print(f"Data type of hemoglobin_g_dl BEFORE conversion: {type(hemoglobin_g_dl)}")
print("Converting now...")

# Cast the string to a float for calculations
hemoglobin_g_dl = float(hemoglobin_g_dl)

print("Conversion completed")
print(hemoglobin_g_dl)
# Output: 14.5
print(f"Data type of hemoglobin_g_dl AFTER conversion: {type(hemoglobin_g_dl)}")
# Output: Data type of hemoglobin_g_dl: <class 'float'>
```

Casting can fail with a `ValueError` if the conversion is impossible (e.g., `int('hello')`).

> **Debug Note:** You will encounter a `ValueError` if you try to convert a string that doesn't look like a number (e.g., `int("N/A")` or `float("pending")`). 

**Data Loss During Casting**
Be extremely cautious when casting a `float` to an `int`. This action can cause a loss of precision.

> **Important:** Casting a float to an integer **truncates** (cuts off) the decimal portion; it does not round. This causes a loss of precision that can be dangerous in a medical context. For example, casting a dosage of `1.9` mL to an integer would result in `1` mL, a potentially critical underdose.
>

```python
# A precise lab value
hemoglobin_level_float = 14.99999999

# Casting to an integer truncates the decimal
hemoglobin_level_truncated = int(hemoglobin_level_float)

print(f"Truncated hemoglobin level: {hemoglobin_level_truncated}")
# Output: Truncated hemoglobin level: 14
```

## 5. Practical Applications

*   **Patient Record Management:** In any Electronic Health Record (EHR) system, data types are fundamental. `str` is used for patient names and MRNs, `int` for age or number of dependents, `float` for height and weight, and `bool` to flag conditions like allergies (`is_allergic = True`) or consent status (`has_signed_consent = True`). Using the wrong type could corrupt records.
*   **Clinical Lab Data Analysis:** When analyzing lab results, `float` is essential for precision (e.g., `glucose = 101.3` mg/dL). If glucose were stored as an `int`, the value would be truncated to `101`, losing information that could be clinically relevant for diagnosing pre-diabetes. Python scripts use these floats to flag values outside normal ranges.
*   **Risk Stratification Models:** Predictive models in healthcare use variables to calculate patient risk scores. A model might use `bool` variables like `is_smoker` or `has_hypertension` alongside `int` (age) and `float` (BMI) variables. The data type of each variable must be correct for the mathematical model to execute properly and produce a valid risk score.
*   **Pharmacogenomics and Dosage Calculation:** Patient-specific drug dosages are often calculated based on precise factors. A `float` is used for the dosage itself (e.g., `1.25` mg), while a `str` might be used to identify a specific genetic variant (e.g., `variant_id = "rs1801133"`) that influences drug metabolism. Incorrect data types would make these personalized calculations impossible.
